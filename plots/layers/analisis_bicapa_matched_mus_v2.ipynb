# §5.4 — Bicapa estratificada a $\mu_s$ igualado: dos reglas de profundidad en un mismo cono

**Qué distingue a esta campaña.** Las cuatro corridas (`NORMAL`/`INVERSE` $\times$ `PCIR`/`PLIN`)
pertenecen a la familia `matched_mus`: la fracción de volumen de la especie grande se deriva de
$\phi_{\rm L}=4a_{\rm L}\mu_s^{\rm ref}/(3Q_{\rm L})$ de modo que **$\mu_s$ es idéntico en las dos
capas**. Sin absorción, esto implica

$$\ell_s^{\rm sup}=\ell_s^{\rm inf}\equiv\ell_s \quad\text{(común a las dos capas y a los dos órdenes)},
\qquad \frac{\ell^*_{\rm L}}{\ell^*_{\rm S}}=\frac{1-g_{\rm S}}{1-g_{\rm L}}\simeq 1.8 .$$

El único contraste es la **anisotropía**. Eso es lo que hace a esta tanda distinta de §5.3 (donde
$\phi$ era fija y $\ell_s$, $\ell^*$ cambiaban a la vez, con contraste 5.09): allí no se podía separar
cuál de las dos longitudes gobernaba cada observable. Aquí sí, porque a **igual $z$ absoluto**

| | $d/\ell_s$ | $d/\ell^*_{\rm ent}$ |
|---|---|---|
| `normal` (entra la pequeña) | $z/\ell_s$ | $z/\ell^*_{\rm S}$ |
| `inverse` (entra la grande) | $z/\ell_s$ **— el mismo número** | $z/\ell^*_{\rm L}$ **— difiere en 1.8** |

Un observable que colapse contra una de las dos columnas y no contra la otra identifica su longitud
de sondeo **sin ningún parámetro ajustado**. Ese es el eje del notebook.

---

## La tesis del capítulo, en una línea

El cono de la bicapa lleva **dos reglas de profundidad simultáneas y de escalas muy distintas**:

- **Regla corta — el ápice.** $\rho=2-\eta(0)$ es un observable de *un solo vértice* (§1.3 del
  documento de fundamentos). Su peso de dispersión simple es, con $\ell_s$ común,
  $$S(d)\;\propto\;\tfrac12\big[p_{\rm sup}(\pi)\,(1-e^{-2d/\ell_s})+p_{\rm inf}(\pi)\,e^{-2d/\ell_s}\big],$$
  de modo que $X\equiv\rho/(1-\rho)=S/M$ progresa como $1-e^{-2d/\ell_s}$: **escala $\ell_s/2$, idéntica
  en los dos órdenes, con el signo del cambio invertido** (porque $p_{\rm sup}$ y $p_{\rm inf}$ se
  intercambian). Cero parámetros libres.
- **Regla larga — el ancho y la dinámica.** El ancho angular y el cono con compuerta temporal son
  observables de transporte: colapsan contra $d/\ell^*_{\rm ent}$, y el quiebre temporal obedece
  $t^*=\tfrac32\,(d/\ell^*_{\rm ent})^2\,\tau^*_{\rm ent}$ con plateau
  $R_\infty=\sqrt{\ell^*_{\rm ent}/\ell^*_{\rm fondo}}$, que **cambia de signo** al invertir el orden.

De ahí sale la parte metrológica: dos rangos de profundidad complementarios ($d\lesssim2\ell_s$ por el
ápice, $d\sim\ell^*$ por el tiempo) y la inversión del medio enterrado vía $R_\infty\to\ell^*_{\rm fondo}
\to g_{\rm fondo}\to a$ (posible sólo porque $\mu_s$ está igualado y $\ell_s$ es conocido).

## 1 — Utilidades de reducción

Tres cambios respecto del pipeline de §5.3, todos motivados por problemas concretos observados allí:

1. **Resolución de canales por física, no por etiqueta.** El nombre del canal difiere entre bases
   `linear` y `circular` y ha habido intercambios de etiqueta. En vez de fiarse del nombre, el canal
   que preserva la helicidad se identifica como **aquel cuyo ápice está más cerca de 2 en los
   controles homogéneos** (Prop. 1.1: $\eta_+(0)=2$ protegido por reciprocidad), y el canal
   co-polarizado lineal como el de mayor ápice. La elección se imprime para inspección y se puede
   forzar con `CHANNEL_OVERRIDE`.
2. **Envolvente monótona antes de invertir niveles.** $q_x$ se obtiene sobre
   $\min$-acumulado del exceso normalizado. Un cono es monótonamente decreciente; imponerlo elimina
   los cruces espurios por ruido sin tocar la señal. Se aplica igual a datos, controles y familia
   sintética, así que cualquier sesgo residual se cancela en razones y progresos.
3. **Todo en $q$**, no en mrad. Como el ancla angular es común a las cuatro campañas, $q$ es
   literalmente el mismo índice de bin y las restas entre campañas no requieren interpolación.

```python
# ===========================================================================
# 1a -- resolucion de canales, stitching, linea base
# ===========================================================================
CHANNEL_OVERRIDE = {}      # p.ej. {"circular": "same", "linear": "co"} para forzar


def _basis_for(pol):
    return circular if pol == "circular" else linear


def _chan_names(prof):
    c = prof.coherent
    try:
        return list(c.keys())
    except AttributeError:
        return [a for a in dir(c) if not a.startswith("_")]


def _chan_get(prof, name, which="coherent"):
    obj = getattr(prof, which)
    try:
        return np.asarray(obj[name], float)
    except (TypeError, KeyError):
        return np.asarray(getattr(obj, name), float)


def raw_profiles(sweep, key, pol, channel, time_index=0):
    """(p1, p2) crudos de las dos ventanas angulares, un solo corte azimutal."""
    sw = sweep[key]
    b = _basis_for(pol)
    p1 = cbs_profiles(sw.processed_cbs("farfield_cbs_1"), basis=b,
                      time_index=time_index, reduce=phi_cut(CONV["phi_main"]))
    p2 = cbs_profiles(sw.processed_cbs("farfield_cbs_2"), basis=b,
                      time_index=time_index, reduce=phi_cut(CONV["phi_main"]))
    return p1, p2


def stitch(th1, y1, th2, y2, key="", warn=0.05):
    """Une ventana fina y cola; avisa si el solape no concuerda."""
    th_1 = th1[-1]
    ov = (th2 >= 0.9 * th_1) & (th2 <= th_1)
    if ov.any():
        resid = np.abs(y2[ov] - np.interp(th2[ov], th1, y1))
        if np.max(resid) > warn:
            print(f"[stitch WARN] {key}: max desacuerdo en solape = {np.max(resid):.3f}")
    keep2 = th2 > th_1
    return (np.concatenate([th1, th2[keep2]]),
            np.concatenate([y1, y2[keep2]]))


def tail_baseline(q, y, q_min=CONV["q_min_baseline"], n_min=5, frac=0.85):
    """Mediana de la cola en q comun. Si la ventana no alcanza, usa el 15 % final."""
    q, y = np.asarray(q, float), np.asarray(y, float)
    m = q > q_min
    if np.count_nonzero(m) >= n_min:
        return float(np.median(y[m]))
    m = q >= frac * np.nanmax(q)
    if np.count_nonzero(m) < n_min:
        return float(np.median(y[np.argsort(q)[-n_min:]]))
    return float(np.median(y[m]))
```

```python
# ===========================================================================
# 1b -- perfiles: una replica y sumado sobre replicas
# ===========================================================================
def profile_one(sweep, key, pol, channel, l_anchor, time_index=0):
    """(q, eta) de UNA replica; linea base llevada a 1.
    Se usa SOLO para barras de error de escalares (la unidad estadistica
    independiente es la replica, no el bin)."""
    p1, p2 = raw_profiles(sweep, key, pol, channel, time_index)
    E1 = (_chan_get(p1, channel, "coherent") + eps) / (_chan_get(p1, channel, "incoherent") + eps)
    E2 = (_chan_get(p2, channel, "coherent") + eps) / (_chan_get(p2, channel, "incoherent") + eps)
    th, E = stitch(np.asarray(p1.theta), E1, np.asarray(p2.theta), E2, key)
    q = k * l_anchor * th
    return q, E - tail_baseline(q, E) + 1.0


def profile_sum(sweep, keys, pol, channel, l_anchor, time_index=0):
    """(q, eta) por cociente de intensidades SUMADAS sobre replicas.
    Estimador correcto del valor central: sum(C)/sum(I), no el promedio de
    cocientes por replica (sesgo de estimador de razon, ec. 17-18)."""
    Ic = Ii = th1 = th2 = None
    for key in keys:
        p1, p2 = raw_profiles(sweep, key, pol, channel, time_index)
        c1, c2 = _chan_get(p1, channel), _chan_get(p2, channel)
        b1, b2 = _chan_get(p1, channel, "incoherent"), _chan_get(p2, channel, "incoherent")
        if Ic is None:
            Ic = [np.zeros_like(c1), np.zeros_like(c2)]
            Ii = [np.zeros_like(b1), np.zeros_like(b2)]
            th1, th2 = np.asarray(p1.theta), np.asarray(p2.theta)
        Ic[0] += c1; Ic[1] += c2; Ii[0] += b1; Ii[1] += b2
    keep2 = th2 > th1[-1]
    th = np.concatenate([th1, th2[keep2]])
    Ct = np.concatenate([Ic[0], Ic[1][keep2]])
    It = np.concatenate([Ii[0], Ii[1][keep2]])
    E = (Ct + eps) / (It + eps)
    q = k * l_anchor * th
    return q, E - tail_baseline(q, E) + 1.0
```

```python
# ===========================================================================
# 1c -- escalares robustos a un perfil de dos escalas
# ===========================================================================
def apex_value(q, E, n=None, mode=None):
    """Valor del apice eta(0).

    El pico CBS es NO analitico en theta=0: eta ~ eta(0) - beta*q + O(q^2)
    (cuspide triangular, ec. 12). Promediar los primeros bins subestima el
    apice, y lo subestima MAS para conos estrechos -> mete una dependencia
    espuria con l* en toda cantidad normalizada por eta(0). Extrapolar la recta
    de la cuspide a q=0 elimina ese sesgo a primer orden para CUALQUIER forma
    de linea, y promedia ruido sobre n bins. La celda de calibracion de abajo
    fija n comparando sesgo (piso sintetico) contra ruido (dispersion entre
    replicas)."""
    n = CONV["n_apex"] if n is None else n
    mode = CONV["apex_mode"] if mode is None else mode
    E = np.asarray(E, float)
    if mode == "median":
        return float(np.median(E[:n]))
    p = np.polyfit(np.asarray(q, float)[:n], E[:n], 1)
    return float(p[1])


def excess_norm(q, E, monotone=None):
    """(eta-1)/(eta(0)-1), con envolvente monotona decreciente opcional.
    Un cono es monotonamente decreciente; imponerlo elimina cruces espurios por
    ruido sin tocar la senal, y se aplica igual a datos, controles y familia
    sintetica, asi que cualquier sesgo residual se cancela en razones."""
    monotone = CONV["monotone_levels"] if monotone is None else monotone
    E0 = apex_value(q, E)
    if not np.isfinite(E0) or E0 <= 1.0:
        return None
    y = (np.asarray(E, float) - 1.0) / (E0 - 1.0)
    return np.minimum.accumulate(y) if monotone else y


def level_q(q, E, x):
    """q donde el exceso normalizado cae a la fraccion x del apice."""
    y = excess_norm(q, E)
    if y is None:
        return np.nan
    b = np.where(y < x)[0]
    b = b[b > 0]
    if len(b) == 0:
        return np.nan
    i = int(b[0])
    if y[i] == y[i - 1]:
        return np.nan
    return float(q[i - 1] + (x - y[i - 1]) * (q[i] - q[i - 1]) / (y[i] - y[i - 1]))


def shape_invariant(q, E):
    """S = q_0.25 / q_0.75. Razon de dos escalas del MISMO perfil, luego es
    independiente de la escala: cualquier medio con un unico l* da el mismo S.
    Leerlo como 'dos reglas del mismo cono que en un medio efectivo coinciden'."""
    a, b = level_q(q, E, CONV["x_lo"]), level_q(q, E, CONV["x_hi"])
    return float(a / b) if (np.isfinite(a) and np.isfinite(b) and b > 0) else np.nan


def scalars_one(sweep, key, pol, channel, l_anchor):
    q, E = profile_one(sweep, key, pol, channel, l_anchor)
    e0 = apex_value(q, E)
    return [e0, 2.0 - e0, level_q(q, E, 0.5), shape_invariant(q, E)]


def scalars_stats(sweep, keys, pol, channel, l_anchor):
    """Media y semi-intervalo t al 95 % sobre replicas de (eta0, rho, q_1/2, S).
    La unidad estadistica independiente es la REPLICA, nunca el bin: los bins
    angulares comparten los mismos caminos simulados y estan correlacionados."""
    A = np.asarray([scalars_one(sweep, kk, pol, channel, l_anchor) for kk in keys], float)
    n = len(A)
    return A.mean(0), t_ci(n) * A.std(0, ddof=1) / np.sqrt(n)


# --- forma normalizada y dispersion del colapso (test de auto-similitud) ----
U_GRID = np.linspace(0.15, 4.0, 200)


def normalized_shape(q, E, u_grid=U_GRID):
    """f(u) = [eta-1]/[eta(0)-1] contra u = q/q_(1/2). Elimina amplitud y escala:
    toda familia de UNA sola escala colapsa exactamente, sea cual sea la forma."""
    y = excess_norm(q, E)
    q_h = level_q(q, E, 0.5)
    if y is None or not np.isfinite(q_h) or q_h <= 0:
        return np.full_like(u_grid, np.nan)
    return np.interp(u_grid, np.asarray(q, float) / q_h, y)


def collapse_spread(curves):
    Cm = np.vstack(curves)
    return float(np.sqrt(np.nanmean((Cm - np.nanmean(Cm, axis=0)) ** 2)))


# --- forma de linea AWM (piso sintetico y ajuste de cono unico) -------------
def cone_shape(qq, b=B_MILNE):
    """Cono difusivo AWM normalizado, C(0)=1 (ec. 10 del documento base)."""
    qq = np.maximum(np.asarray(qq, float), 1e-12)
    return (1.0 / (1.0 + 2.0 * b)) / (1.0 + qq) ** 2 * \
           (1.0 + (1.0 - np.exp(-2.0 * b * qq)) / qq)


# --- ajuste con la forma de linea MEDIDA, abscisa estirada -----------------
#   Sustituye al ajuste AWM. Motivo: sobre los controles homogeneos el ajuste
#   AWM deja RMS ~ 1.3e-2, que NO es ruido sino el fallo de AWM al describir el
#   cono polarizado medido (no-universalidad del prefactor, 5.1). Usar como
#   patron el propio control medido convierte ese piso en ruido estadistico y
#   devuelve poder al test de "una sola escala".
def stretched(q, q_ref, E_ref, A, s):
    """1 + A * [eta_ref(q/s) - 1]: el control medido, reescalado en angulo."""
    return 1.0 + A * np.interp(np.asarray(q, float) / s, q_ref, E_ref - 1.0,
                               left=E_ref[0] - 1.0, right=0.0)


def fit_stretched(q, E, q_ref, E_ref, q_max=None):
    """Mejor descripcion de (q,E) por UNA sola escala, tomando como forma de
    linea el control medido. Devuelve (A, s, RMS)."""
    q_max = CONV["q_fit"] if q_max is None else q_max
    m = np.asarray(q, float) <= q_max
    f = lambda qq, A, s: stretched(qq, q_ref, E_ref, A, s)
    a0 = (apex_value(q, E) - 1.0) / max(apex_value(q_ref, E_ref) - 1.0, 1e-6)
    popt, _ = curve_fit(f, np.asarray(q)[m], np.asarray(E)[m],
                        p0=[a0, 1.0], maxfev=40000)
    resid = np.asarray(E)[m] - f(np.asarray(q)[m], *popt)
    return float(popt[0]), float(popt[1]), float(np.sqrt(np.mean(resid ** 2)))


def fit_two_cones(q, E, E_entry, E_back, q_max=None):
    """Peso w de la capa de entrada por minimos cuadrados cerrados sobre
    (y - f_back) = w (f_entry - f_back). Usa los DOS controles medidos."""
    q_max = CONV["q_fit"] if q_max is None else q_max
    m = np.asarray(q, float) <= q_max
    ft, fb, y = E_entry - 1.0, E_back - 1.0, np.asarray(E, float) - 1.0
    dd = (ft - fb)[m]
    w = float(np.sum((y - fb)[m] * dd) / np.sum(dd ** 2))
    resid = (y - fb - w * (ft - fb))[m]
    return w, float(np.sqrt(np.mean(resid ** 2)))


def on_ref_grid(q, E, q_ref):
    """Lleva (q,E) a la grilla del control. Las 4 campanias NO comparten
    fingerprint (l* difiere ~0.1 % entre ellas), asi que esto es obligatorio."""
    return E if (len(q) == len(q_ref) and np.allclose(q, q_ref)) else np.interp(q_ref, q, E)
```

## 2 — Carga de las cuatro campañas

Cada campaña trae, además de los 12 espesores $\times$ 5 réplicas, **dos controles semi-infinitos
explícitos** (`single_layer_small`, `single_layer_large`) con la misma grilla y las mismas réplicas.
Es una mejora sobre §5.3, donde los controles eran los extremos del propio barrido: ahora el límite
homogéneo es una medición independiente, no una extrapolación.

Las clasificaciones se leen de `params_flat` (`layer_kind`, `single_layer_species`, `order`), y el
orden se **re-deriva** de los radios como verificación cruzada. El `grid_fingerprint` debe coincidir
en las cuatro campañas: es la licencia para restar perfiles bin a bin.

```python
# ===========================================================================
# 2a -- estructuras y lector de campania
# ===========================================================================
@dataclass
class Layer:
    z: float                      # espesor absoluto de la capa de entrada [um]
    m_star: float                 # d / l*_entrada   <- regla LARGA
    m_s: float                    # d / l_s          <- regla CORTA (comun a ordenes)
    reachable: bool
    keys: list = field(default_factory=list)


@dataclass
class Campaign:
    tag: str; order: str; pol: str
    sweep: object
    r_entry: float; r_back: float
    l_entry: float; l_back: float          # l* de entrada y de fondo
    ls_common: float                       # l_s (identico en ambas capas)
    g_entry: float; g_back: float
    l_anchor: float                        # ancla angular comun a las 4 campanias
    fingerprint: str
    layers: list
    ctrl: dict                             # 'entry' / 'back' -> lista de keys
    dt_sim: float; t_max_sim: float; n_time: int
    channel: str = ""

    @property
    def m_star_arr(self): return np.array([L.m_star for L in self.layers])
    @property
    def m_s_arr(self):    return np.array([L.m_s for L in self.layers])
    @property
    def z_arr(self):      return np.array([L.z for L in self.layers])
    @property
    def tau_entry(self):
        """tau*_entrada en unidades de simulacion (camino geometrico en um)."""
        return self.l_entry
    @property
    def R_inf(self):
        """Plateau tardio del cociente de anchos: sqrt(l*_ent / l*_fondo)."""
        return float(np.sqrt(self.l_entry / self.l_back))


def _get(p, names, default=np.nan):
    """Lee el primer nombre presente; tolera aplanado por punto o guion bajo."""
    for nm in names:
        for cand in (nm, nm.replace(".", "_"), nm.replace("_", ".")):
            if cand in p:
                try:
                    return float(p[cand])
                except (TypeError, ValueError):
                    return p[cand]
    return default


def _get_str(p, names, default=""):
    v = _get(p, names, default)
    return v if isinstance(v, str) else default


def _find_g(p, which):
    """g de la capa 'top'/'bot': prueba nombres directos y luego busca en las
    claves aplanadas de dq_top / dq_bot."""
    v = _get(p, [f"dq_{which}.anisotropy_g", f"g_{which}", f"anisotropy_g_{which}"])
    if isinstance(v, float) and np.isfinite(v):
        return v
    for kk in p:
        if which in kk and "anisotropy" in kk:
            try:
                return float(p[kk])
            except (TypeError, ValueError):
                pass
    return np.nan
```

```python
# ===========================================================================
# 2b -- carga
# ===========================================================================
def load_campaign(order, pol, folder):
    sw = load_sweep(folder, base_path=Path(BASE_PATH))

    by_z, ctrl = {}, {"small": [], "large": []}
    p_any = None
    for key in sw.keys():
        p = sw[key].params_flat
        p_any = p_any or p
        kind = _get_str(p, ["layer_kind"])
        if kind == "stratified_one_layer" or "single_layer" in key:
            spec = _get_str(p, ["single_layer_species"])
            if not spec:
                spec = "small" if "small" in key else "large"
            ctrl[spec].append(key)
            continue
        z = _get(p, ["z_interface"])
        if not (isinstance(z, float) and np.isfinite(z)):
            mm = re.search(r"z_interface_([0-9.]+)__rep(\d+)", key)
            if not mm:
                print(f"[{order}/{pol}] [skip] {key}")
                continue
            z = float(mm.group(1))
        by_z.setdefault(round(z, 6), []).append(key)

    p0 = sw[sorted(by_z[sorted(by_z)[0]])[0]].params_flat
    r_top, r_bot = _get(p0, ["radius_top"]), _get(p0, ["radius_bot"])
    l_top, l_bot = _get(p0, ["l_star_top"]), _get(p0, ["l_star_bot"])
    ls_top, ls_bot = _get(p0, ["l_s_top"]), _get(p0, ["l_s_bot"])
    def _g_or(val, ls_, l_):
        """g archivado si existe; si no, exacto desde g = 1 - l_s/l*."""
        try:
            v = float(val)
            if np.isfinite(v):
                return v
        except (TypeError, ValueError):
            pass
        return 1.0 - ls_ / l_
    g_top = _g_or(_find_g(p0, "top"), ls_top, l_top)
    g_bot = _g_or(_find_g(p0, "bot"), ls_bot, l_bot)
    l_ang = _get(p0, ["l_star_angle_anchor"], min(l_top, l_bot))
    fp = _get_str(p0, ["grid_fingerprint"])
    dt = _get(p0, ["d_time", "time_grid.dt_sim", "dt_sim"])
    tmax = _get(p0, ["t_max", "time_grid.t_max_sim", "t_max_sim"])
    n_t = int(round(tmax / dt)) if np.isfinite(dt) and dt > 0 else 0

    # verificacion del invariante de la familia: mu_s igualado -> l_s comun
    rel = abs(ls_top - ls_bot) / ls_top
    if rel > 1e-4:
        print(f"[{order}/{pol}] AVISO: l_s NO comun (desviacion relativa {rel:.2e}). "
              f"La regla corta de MM3 supone l_s comun.")
    ls = 0.5 * (ls_top + ls_bot)

    # orden re-derivado de los radios, contra el campo archivado
    order_arch = _get_str(p0, ["order"])
    order_der = "inverse" if r_top > r_bot else "normal"
    if order_arch and order_arch != order_der:
        print(f"[{order}/{pol}] AVISO: order archivado '{order_arch}' != derivado "
              f"'{order_der}' (se usa el derivado)")

    layers = []
    for z in sorted(by_z):
        pz = sw[sorted(by_z[z])[0]].params_flat
        reach = bool(_get(pz, ["reachable_paired"], True))
        layers.append(Layer(z=z, m_star=z / l_top, m_s=z / ls,
                            reachable=reach, keys=sorted(by_z[z])))

    entry_spec = "large" if r_top > r_bot else "small"
    back_spec = "small" if entry_spec == "large" else "large"

    return Campaign(tag=f"{order}/{pol}", order=order_der, pol=pol, sweep=sw,
                    r_entry=r_top, r_back=r_bot, l_entry=l_top, l_back=l_bot,
                    ls_common=ls, g_entry=g_top, g_back=g_bot,
                    l_anchor=l_ang, fingerprint=fp, layers=layers,
                    ctrl={"entry": sorted(ctrl[entry_spec]),
                          "back": sorted(ctrl[back_spec])},
                    dt_sim=dt, t_max_sim=tmax, n_time=n_t)


CAMP = {}
for (order, pol), folder in FOLDERS.items():
    try:
        CAMP[(order, pol)] = load_campaign(order, pol, folder)
        c = CAMP[(order, pol)]
        print(f"[{order:7s}/{pol:8s}] OK  {len(c.layers)} espesores, "
              f"controles: entrada {len(c.ctrl['entry'])} rep, fondo {len(c.ctrl['back'])} rep")
    except Exception as err:
        print(f"[{order:7s}/{pol:8s}] no disponible: {err}")

assert CAMP, "ninguna campania cargada: revisa BASE_PATH y FOLDERS"
LREF = list(CAMP.values())[0].l_anchor          # ancla angular comun
```

```python
# ===========================================================================
# 2c -- tabla del sistema y chequeos de comparabilidad
# ===========================================================================
fps = {c.fingerprint for c in CAMP.values() if c.fingerprint}
print(f"grid_fingerprint distintos: {len(fps)}  -> "
      f"{'restas bin a bin LEGITIMAS' if len(fps) <= 1 else 'AVISO: grillas distintas, interpola antes de restar'}")
for fp in fps:
    print(f"   {fp}")

rows = []
for (order, pol), c in CAMP.items():
    rows.append(dict(order=order, pol=pol,
                     r_entry_nm=1e3 * c.r_entry, r_back_nm=1e3 * c.r_back,
                     l_s_um=c.ls_common,
                     l_star_entry=c.l_entry, l_star_back=c.l_back,
                     g_entry=c.g_entry, g_back=c.g_back,
                     contrast=c.l_back / c.l_entry, R_inf=c.R_inf,
                     dt_sim=c.dt_sim, t_max_sim=c.t_max_sim, n_time=c.n_time))
SYS = pd.DataFrame(rows)
display(SYS.round(4))

c0 = list(CAMP.values())[0]
print(f"\nl_s comun            = {c0.ls_common:.3f} um")
print(f"l* pequena / grande  = {min(c0.l_entry, c0.l_back):.3f} / {max(c0.l_entry, c0.l_back):.3f} um"
      f"   -> contraste {max(c0.l_entry, c0.l_back)/min(c0.l_entry, c0.l_back):.3f}")
print(f"ancla angular LREF   = {LREF:.3f} um   (q = k*LREF*theta)")
print("\nespesores del barrido:")
print("   z [um] | d/l_s | d/l*_normal | d/l*_inverse | alcance temporal")
n_common = min(len(c.layers) for c in CAMP.values())
if n_common != len(c0.layers):
    print(f"AVISO: numero de espesores distinto entre campanias "
          f"({[len(c.layers) for c in CAMP.values()]}); se listan los {n_common} comunes")
for i, L in enumerate(c0.layers[:n_common]):
    ms = {}
    for (o, p), c in CAMP.items():
        ms[o] = c.layers[i].m_star
    print(f"  {L.z:8.2f} | {L.m_s:5.2f} | {ms.get('normal', np.nan):11.2f} | "
          f"{ms.get('inverse', np.nan):12.2f} | {'si' if L.reachable else 'NO':>8}")
print("\n  d/l_s es el MISMO numero en las dos campanias (mu_s igualado);")
print("  d/l*_ent difiere por el contraste. Esa asimetria es el experimento.")
```

```python
# ===========================================================================
# 2d -- resolucion de canales POR FISICA (no por etiqueta)
#   circular : el canal que preserva helicidad es el de apice ~ 2 (Prop. 1.1)
#   lineal   : el co-polarizado es el de mayor apice
# ===========================================================================
def resolve_channel(c):
    if c.pol in CHANNEL_OVERRIDE:
        return CHANNEL_OVERRIDE[c.pol]
    key = c.ctrl["entry"][0] if c.ctrl["entry"] else c.layers[-1].keys[0]
    p1, _ = raw_profiles(c.sweep, key, c.pol, None)
    names = _chan_names(p1)
    apex = {}
    for nm in names:
        try:
            q, E = profile_one(c.sweep, key, c.pol, nm, c.l_anchor)
            apex[nm] = apex_value(q, E)
        except Exception:
            continue
    if not apex:
        raise RuntimeError(f"[{c.tag}] no se pudo leer ningun canal de {names}")
    if c.pol == "circular":
        pick = min(apex, key=lambda nm: abs(apex[nm] - 2.0))
    else:
        pick = max(apex, key=lambda nm: apex[nm])
    txt = "  ".join(f"{nm}: eta(0)={v:.4f}" for nm, v in apex.items())
    print(f"[{c.tag}] canales {names}  ->  {txt}   ELEGIDO: '{pick}'")
    return pick


for c in CAMP.values():
    c.channel = resolve_channel(c)

print("\nCanales en uso:", {t: c.channel for t, c in
                            zip([f"{o}/{p}" for (o, p) in CAMP], CAMP.values())})
print("Si alguno no corresponde a lo esperado, fija CHANNEL_OVERRIDE y re-ejecuta.")
```

```python
# ===========================================================================
# 2e -- perfiles estacionarios (una sola vez; todo lo demas los reutiliza)
# ===========================================================================
PROF = {}     # (order, pol, z)             -> (q, eta)
CTRL = {}     # (order, pol, 'entry'|'back')-> (q, eta)

for (o, p), c in CAMP.items():
    for L in c.layers:
        PROF[(o, p, L.z)] = profile_sum(c.sweep, L.keys, c.pol, c.channel, c.l_anchor)
    for which in ("entry", "back"):
        if c.ctrl[which]:
            CTRL[(o, p, which)] = profile_sum(c.sweep, c.ctrl[which], c.pol,
                                              c.channel, c.l_anchor)
print(f"perfiles estacionarios: {len(PROF)} bicapas, {len(CTRL)} controles")

POL_GATE = "circular" if any(p == "circular" for (_, p) in CAMP) else "linear"
Q_GRID = list(CTRL.values())[0][0] if CTRL else list(PROF.values())[0][0]
print(f"grilla en q: {len(Q_GRID)} bins, dq(apice) = {Q_GRID[1]-Q_GRID[0]:.4f}, "
      f"q_max = {Q_GRID[-1]:.1f}")
```

```python
# ===========================================================================
# 2f -- calibracion del estimador de apice: sesgo contra ruido
#   El unico parametro libre del pipeline es n_apex. Se fija ANTES de mirar
#   fisica, con dos numeros que tiran en direcciones opuestas:
#     sesgo : dispersion residual de una familia SINTETICA de una sola escala
#             sobre la grilla REAL. Deberia ser 0; lo que quede es artefacto
#             del estimador y pone el piso del test de auto-similitud (MM2).
#     ruido : dispersion de eta(0) entre replicas de un control homogeneo.
#   Se elige el menor n cuyo sesgo cae por debajo del ruido.
# ===========================================================================
def synthetic_single_scale(q_grid, l_lo, l_hi, n=12, amp=(0.6, 1.0)):
    """Familia AWM de UNA sola escala sobre el rango real de l*, pasada por el
    MISMO pipeline y la MISMA grilla que los datos."""
    fam = []
    for i, ls in enumerate(np.geomspace(l_lo, l_hi, n)):
        A = amp[0] + (amp[1] - amp[0]) * i / max(n - 1, 1)
        fam.append((np.asarray(q_grid, float),
                    1.0 + A * cone_shape(np.asarray(q_grid, float) * ls / LREF)))
    return fam


L_LO = min(min(c.l_entry, c.l_back) for c in CAMP.values())
L_HI = max(max(c.l_entry, c.l_back) for c in CAMP.values())
SYN = synthetic_single_scale(Q_GRID, L_LO, L_HI)

_ctrl_key = next(iter(CTRL))
_c_ref = CAMP[(_ctrl_key[0], _ctrl_key[1])]
_ctrl_keys = _c_ref.ctrl[_ctrl_key[2]]

print(f"rango de l* cubierto por el piso sintetico: {L_LO:.2f}-{L_HI:.2f} um "
      f"({L_HI/L_LO:.2f}x)")
print("\n   n_apex | sesgo (piso sintetico) | ruido (replicas de eta_0) | S sintetico")
_scan = {}
for n_try in (3, 4, 5, 6, 8, 12):
    _old = CONV["n_apex"]
    CONV["n_apex"] = n_try
    bias = collapse_spread([normalized_shape(qq, EE) for qq, EE in SYN])
    Ss = [shape_invariant(qq, EE) for qq, EE in SYN]
    s_var = 100.0 * (np.nanmax(Ss) / np.nanmin(Ss) - 1.0)
    e0s = [apex_value(*profile_one(_c_ref.sweep, kk, _c_ref.pol,
                                   _c_ref.channel, _c_ref.l_anchor))
           for kk in _ctrl_keys]
    noise = float(np.std(e0s, ddof=1))
    CONV["n_apex"] = _old
    _scan[n_try] = (bias, noise, s_var)
    print(f"   {n_try:6d} | {bias:22.2e} | {noise:25.2e} | {s_var:10.2f} %")

_ok = [n for n, (b, nz, _) in _scan.items() if b < nz]
N_APEX_CAL = min(_ok) if _ok else min(_scan, key=lambda n: _scan[n][0])
print(f"\n   -> n_apex = {N_APEX_CAL}  "
      f"({'sesgo por debajo del ruido de replica' if _ok else 'AVISO: sesgo domina en todo el rango'})")
CONV["n_apex"] = N_APEX_CAL
SYN_FLOOR = _scan[N_APEX_CAL][0]
print(f"   piso del procedimiento para MM2: {SYN_FLOOR:.2e}")
print("   (si el sesgo domina, sube N_THETA_1 en la campania o usa apex_mode='median'\n"
      "    y reporta el piso como limite del test, no como resultado)")
```

## MM0 — Compuertas: nada de lo que sigue es interpretable si esto falla

Dos compuertas independientes, ambas sin parámetros:

1. **Reciprocidad (canal circular).** $\eta_+(0)=2$ está protegido por reciprocidad para dispersores
   con simetría de espejo, *cualquiera sea el medio* (Prop. 1.1). Una bicapa no puede romperlo: si
   $\eta_+(0)$ se mueve con el espesor, el problema es de la maquinaria de capas, no física. Es la
   misma función que cumplió $\eta_+(0)=1.9896$ constante en las mezclas de §5.2.
2. **Límites del barrido.** El espesor mínimo ($d/\ell_s\sim10^{-5}$) debe reproducir el **control
   del fondo**, y el máximo debe acercarse al **control de entrada**. Como las dos campañas
   intercambian los papeles, cada límite homogéneo tiene dos realizaciones independientes: si
   coinciden, el pipeline es consistente en los dos órdenes a la vez.

```python
# ===========================================================================
# MM0 -- compuertas de reciprocidad y de limites
# ===========================================================================
apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.26})

# --- (a) reciprocidad: eta(0) vs espesor, canal circular -------------------
GATE = {}
for (o, p), c in CAMP.items():
    if p != "circular":
        continue
    mu, se = zip(*[scalars_stats(c.sweep, L.keys, c.pol, c.channel, c.l_anchor)
                   for L in c.layers])
    e0 = np.array([m[0] for m in mu]); s0 = np.array([s[0] for s in se])
    st = STYLE[o]
    ax1.errorbar(c.m_s_arr + 1e-3, e0, yerr=s0, fmt=st["marker"], ms=4, lw=0.9,
                 color=st["color"], ls=st["ls"], capsize=2, label=st["label"])
    GATE[("recip", o)] = dict(mean=float(np.mean(e0)),
                              spread=float(np.max(e0) - np.min(e0)),
                              max_dev=float(np.max(np.abs(e0 - 2.0))))
    print(f"[MM0 recip {o:7s}] eta_+(0) = {np.mean(e0):.4f} +- {np.std(e0, ddof=1):.4f}"
          f"   rango {e0.min():.4f}-{e0.max():.4f}   |desv. de 2| max = {np.max(np.abs(e0-2)):.4f}")

ax1.axhline(2.0, c="k", lw=0.7, ls=":")
ax1.set_xscale("log")
ax1.set_xlabel(r"$d/\ell_s$")
ax1.set_ylabel(r"$\eta_+(0)$  (helicity preserving)")
ax1.set_title("(a) reciprocity gate", loc="left")
ax1.grid(alpha=0.15, which="both")
ax1.legend(fontsize="x-small", frameon=False)

# --- (b) limites vs controles --------------------------------------------
qv = CONV["q_view"]
for (o, p), c in CAMP.items():
    if p != POL_GATE:
        continue
    st = STYLE[o]
    q_thin, E_thin = PROF[(o, p, c.layers[0].z)]
    q_thick, E_thick = PROF[(o, p, c.layers[-1].z)]
    w = q_thin <= qv
    ax2.plot(q_thin[w], E_thin[w], color=st["color"], lw=1.0, ls="-", alpha=0.9,
             label=rf"{o}: $d\to0$")
    w = q_thick <= qv
    ax2.plot(q_thick[w], E_thick[w], color=st["color"], lw=1.0, ls="--", alpha=0.9,
             label=rf"{o}: $d\to\infty$")
    for which, mk in (("entry", ":"), ("back", "-.")):
        if (o, p, which) not in CTRL:
            continue
        qc, Ec = CTRL[(o, p, which)]
        w = qc <= qv
        ax2.plot(qc[w], Ec[w], color=st["color"], lw=0.7, ls=mk, alpha=0.55)

# --- numeros de la compuerta de limites -----------------------------------
for (o, p), c in CAMP.items():
    if p != POL_GATE:
        continue
    for lim, which in (("thin", "back"), ("thick", "entry")):
        if (o, p, which) not in CTRL:
            continue
        L = c.layers[0] if lim == "thin" else c.layers[-1]
        qd, Ed = PROF[(o, p, L.z)]
        qc, Ec = CTRL[(o, p, which)]
        w = qd <= qv
        d = Ed[w] - np.interp(qd[w], qc, Ec)
        GATE[("limit", o, lim)] = dict(rms=float(np.sqrt(np.mean(d ** 2))),
                                       max=float(np.max(np.abs(d))))
        print(f"[MM0 limit {o:7s} {lim:5s}] vs control '{which}': "
              f"RMS = {np.sqrt(np.mean(d**2)):.3e}   max = {np.max(np.abs(d)):.3e}")

ax2.set_xlabel(r"$q = k\,\ell^*_{\rm anchor}\,\theta$")
ax2.set_ylabel(r"$\eta$")
ax2.set_title(f"(b) sweep limits vs controls [{POL_GATE}]", loc="left")
ax2.grid(alpha=0.15)
ax2.legend(fontsize=6.5, frameon=False, ncol=2, columnspacing=0.8)
fig.savefig(FIGDIR / "MM0_compuertas.pdf", bbox_inches="tight")
plt.show()
```

## MM2 — ¿Admite la bicapa un $\ell^*$ efectivo? Test libre de modelo

Se normaliza cada perfil en amplitud por $\eta(0)-1$ y en ángulo por su propia semianchura,
$\tilde f(u)=[\eta-1]/[\eta(0)-1]$ con $u=q/q_{1/2}$. Si un medio admite una única longitud de
transporte, $\tilde f$ no depende ni de la amplitud ni de la escala: **toda la familia colapsa,
sea cual sea la forma de línea** (Prop. 4.1).

> **Corrección respecto de la primera ejecución.** Antes se agrupaban los ocho controles de las
> **dos** polarizaciones en un solo "control positivo". Eso es inválido: la forma de línea depende
> del canal — es justo lo que estableció §5.1 al descartar un prefactor universal entre canales — y
> aquella dispersión medía la diferencia circular/lineal, no el efecto de $\ell^*$. Por eso el
> control positivo salía **peor** que las bicapas. Todas las referencias se construyen ahora dentro
> de un mismo canal.

Con la corrección, la jerarquía de referencias tiene tres escalones y hay que leerla entera:

1. **Sintético** — familia AWM de una escala sobre la grilla real. Mide el sesgo del procedimiento.
2. **Repetición** — el *mismo* medio homogéneo medido en las dos campañas. Ruido estadístico puro.
3. **Los dos medios homogéneos del mismo canal** — **éste es el piso honesto del test.** Las dos
   especies no comparten forma de línea ni tras normalizar la escala (el canal circular las separa
   un ~15 % en $S$), así que ninguna familia que las mezcle puede colapsar mejor que esto. Compararse
   con el piso sintético sobreestimaría el resultado por dos órdenes de magnitud.

El panel (b) da lo mismo en forma escalar: $S=q_{0.25}/q_{0.75}$ es constante para cualquier medio de
una escala, y la banda gris son los constituyentes — el piso, no el cero.

```python
# ===========================================================================
# MM2 -- auto-similitud, con las referencias construidas POR CANAL
#   Correccion respecto de la version anterior: agrupar los controles de las
#   dos polarizaciones en un solo "control positivo" es invalido. La forma de
#   linea depende del canal (5.1: no hay prefactor universal entre canales),
#   asi que esa dispersion medía la diferencia circular/lineal y no el efecto
#   de l*. Todas las referencias se construyen ahora DENTRO de un canal.
# ===========================================================================
def species_of(c, which):
    small_is_entry = c.r_entry < c.r_back
    if which == "entry":
        return "small" if small_is_entry else "large"
    return "large" if small_is_entry else "small"


CT = {}                       # (pol, medio) -> lista de (q, eta) realizaciones
for (o, p), c in CAMP.items():
    for which in ("entry", "back"):
        if (o, p, which) in CTRL:
            CT.setdefault((p, species_of(c, which)), []).append(CTRL[(o, p, which)])

SHAPES, SPREAD = {}, {}
SHAPES["synthetic"] = [normalized_shape(qq, EE) for qq, EE in SYN]
SPREAD["synthetic"] = collapse_spread(SHAPES["synthetic"])

# --- piso 1: repeticion del MISMO medio en las dos campanias -> ruido puro --
for p in pols:
    reps = []
    for med in ("small", "large"):
        cur = CT.get((p, med), [])
        if len(cur) >= 2:
            reps.append(collapse_spread([normalized_shape(*x) for x in cur]))
    if reps:
        SPREAD[f"repeat/{p}"] = float(np.mean(reps))

# --- piso 2: los DOS medios homogeneos del mismo canal ---------------------
#   Es el piso HONESTO del test: si las dos especies no comparten forma de
#   linea, ninguna familia que las mezcle puede colapsar mejor que esto.
for p in pols:
    cur = [x for med in ("small", "large") for x in CT.get((p, med), [])]
    if len(cur) >= 2:
        SHAPES[f"controls/{p}"] = [normalized_shape(*x) for x in cur]
        SPREAD[f"controls/{p}"] = collapse_spread(SHAPES[f"controls/{p}"])

# --- las bicapas -----------------------------------------------------------
for (o, p), c in CAMP.items():
    SHAPES[f"{o}/{p}"] = [normalized_shape(*PROF[(o, p, L.z)]) for L in c.layers]
    SPREAD[f"{o}/{p}"] = collapse_spread(SHAPES[f"{o}/{p}"])

print("[MM2] dispersion del colapso, comparada SIEMPRE dentro del mismo canal")
print(f"      {'referencia':22s} {'dispersion':>10s}   lectura")
print(f"      {'sintetico (una escala)':22s} {SPREAD['synthetic']:10.2e}   sesgo del procedimiento")
for p in pols:
    if f"repeat/{p}" in SPREAD:
        print(f"      {'repeticion, '+p:22s} {SPREAD[f'repeat/{p}']:10.2e}   ruido estadistico")
    if f"controls/{p}" in SPREAD:
        print(f"      {'2 medios homog., '+p:22s} {SPREAD[f'controls/{p}']:10.2e}   PISO del test (formas distintas)")
    for o in orders:
        if (o, p) in CAMP:
            v = SPREAD[f"{o}/{p}"]
            base = SPREAD.get(f"controls/{p}", np.nan)
            print(f"      {'bicapa '+o+', '+p:22s} {v:10.2e}   {v/base:.1f}x el piso del canal")

VERDICT_MM2 = {}
for p in pols:
    base = SPREAD.get(f"controls/{p}")
    if base is None:
        continue
    rs = [SPREAD[f"{o}/{p}"] / base for o in orders if (o, p) in CAMP]
    VERDICT_MM2[p] = float(np.min(rs))
    ok = VERDICT_MM2[p] > 1.5
    print(f"      -> {p}: las bicapas dispersan {np.min(rs):.1f}-{np.max(rs):.1f}x el piso"
          f"  ({'NO admiten una sola escala' if ok else 'NO discriminable de una sola escala'})")

# --- escalares por replica (los usan MM3, MM4, MM7 y la tabla) -------------
SCAL = {}
for (o, p), c in CAMP.items():
    mu, se = zip(*[scalars_stats(c.sweep, L.keys, c.pol, c.channel, c.l_anchor)
                   for L in c.layers])
    MU, SE = np.asarray(mu), np.asarray(se)
    SCAL[(o, p)] = dict(m_s=c.m_s_arr, m_star=c.m_star_arr, z=c.z_arr,
                        eta0=MU[:, 0], s_eta0=SE[:, 0], rho=MU[:, 1], s_rho=SE[:, 1],
                        qh=MU[:, 2], s_qh=SE[:, 2], S=MU[:, 3], s_S=SE[:, 3])

S_BAND = {}
for p in pols:
    vals = [shape_invariant(*x) for med in ("small", "large") for x in CT.get((p, med), [])]
    vals = np.asarray([v for v in vals if np.isfinite(v)], float)
    if len(vals):
        S_BAND[p] = (float(vals.min()), float(vals.max()), float(vals.mean()))
        print(f"[MM2] S de los homogeneos en {p}: {vals.mean():.3f} "
              f"(rango {vals.min():.3f}-{vals.max():.3f}, "
              f"{100*(vals.max()-vals.min())/vals.mean():.1f} %) "
              f"-- ese % ES el piso, no cero")

# --- figura ----------------------------------------------------------------
apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.26})

for sh in SHAPES["synthetic"]:
    ax1.plot(U_GRID, sh, color="0.8", lw=0.6, zorder=1)
ax1.plot([], [], color="0.8", lw=0.8, label="synthetic single scale")
for sh in SHAPES.get(f"controls/{POL_GATE}", []):
    ax1.plot(U_GRID, sh, color="k", lw=1.0, ls=":", zorder=3)
if f"controls/{POL_GATE}" in SHAPES:
    ax1.plot([], [], color="k", lw=1.0, ls=":", label="homogeneous controls (floor)")
for o in orders:
    if (o, POL_GATE) not in CAMP:
        continue
    st = STYLE[o]
    for sh in SHAPES[f"{o}/{POL_GATE}"]:
        ax1.plot(U_GRID, sh, color=st["color"], lw=0.6, alpha=0.7, zorder=4)
    ax1.plot([], [], color=st["color"], lw=0.9, label=f"bilayer, {o}")
ax1.set_xlabel(r"$u = q/q_{1/2}$"); ax1.set_ylabel(r"$\tilde f(u)$")
ax1.set_title(f"(a) self-similarity collapse [{POL_GATE}]", loc="left", fontsize="small")
ax1.set_xlim(0.15, 3.0); ax1.grid(alpha=0.15)
ax1.legend(fontsize=6, frameon=False)

if POL_GATE in S_BAND:
    lo, hi, _ = S_BAND[POL_GATE]
    ax2.axhspan(lo, hi, color="0.85", zorder=0, label="homogeneous constituents")
for o in orders:
    if (o, POL_GATE) not in SCAL:
        continue
    D, st = SCAL[(o, POL_GATE)], STYLE[o]
    ax2.errorbar(D["m_s"] + 1e-3, D["S"], yerr=D["s_S"], fmt=st["marker"], ms=4,
                 lw=0.9, ls=st["ls"], color=st["color"], capsize=2, label=st["label"])
    if POL_GATE in S_BAND:
        _, _, mid = S_BAND[POL_GATE]
        dev = 100 * (D["S"] / mid - 1)
        i = int(np.nanargmax(np.abs(dev)))
        print(f"[MM2 {o:7s}] desviacion max de S = {dev[i]:+.1f} % en d/l_s = {D['m_s'][i]:.2f}"
              f"   (piso de constituyentes {100*(S_BAND[POL_GATE][1]-S_BAND[POL_GATE][0])/mid:.0f} %)")
ax2.set_xscale("log")
ax2.set_xlabel(r"$d/\ell_s$"); ax2.set_ylabel(r"$S = q_{0.25}/q_{0.75}$")
ax2.set_title("(b) shape invariant", loc="left", fontsize="small")
ax2.grid(alpha=0.15, which="both"); ax2.legend(fontsize=6, frameon=False)
fig.savefig(FIGDIR / "MM2_autosimilitud.pdf", bbox_inches="tight")
plt.show()
```

## MM3 — Lo que la familia $\mu_s$ igualado sí mide, y lo que no

Tres cosas distintas, en orden decreciente de solidez.

**(i) El ancho estacionario sigue a $\ell^*$, no a $\ell_s$.** En los dos medios **homogéneos**,
$\ell_s$ es idéntico por construcción y sólo difiere $\ell^*$. Si el ancho midiera $\ell_s$, los dos
controles darían el mismo $q_{1/2}$. La predicción es
$q_{1/2}^{\rm gra}/q_{1/2}^{\rm peq}=\ell^*_{\rm peq}/\ell^*_{\rm gra}$, sin parámetros. Es el único
sitio donde la familia entrega limpiamente lo que prometía — y lo hace sin necesidad de bicapa.

**(ii) El ápice sale del intervalo de sus constituyentes, en los dos órdenes.** $\rho=2-\eta(0)$ es
un observable de un solo vértice y en una mezcla era exactamente aditivo, luego acotado por los dos
constituyentes [Eq. 39: un promedio convexo]. En la bicapa **excursiona fuera del intervalo**, y aquí
en **ambos** órdenes con signos opuestos y el extremo en el mismo $d/\ell_s$.

Esto es más fuerte que la reproducción de §5.3. Allí el sobreimpulso aparecía sólo en un orden y se
atribuyó a que sólo esa campaña tenía la capa de entrada ópticamente delgada. Con $\mu_s$ igualado,
**las dos capas de entrada son ópticamente idénticas a igual $z$**, y en efecto las dos excursionan:
el mecanismo (capa delgada que aporta dispersión simple sin aportar difusión) pasa de plausible a
contrastado. Y un promedio ponderado no puede salir de su propio intervalo: es una demostración de
ausencia de medio efectivo independiente del colapso.

**(iii) El test de regla — y por qué esta campaña no lo decide.** La idea de diseño era: a igual $z$,
$d/\ell_s$ coincide entre órdenes y $d/\ell^*_{\rm ent}$ difiere por el contraste, así que el
observable colapsará en la normalización correcta. Los paneles (b) y (c) lo muestran con el peso
$w(d)$ de la capa de entrada. **El resultado es que no discrimina:** $q_{1/2}$ y $w$ prefieren reglas
opuestas. El poder del test *es* el contraste, y con 1.82 se lo comen la diferencia de forma entre
constituyentes y el ruido. La celda calcula el veredicto y lo dice; no se cita una preferencia que
los datos no sostienen.

> La predicción $P(X)=1-e^{-2d/\ell_s}$ de la versión anterior está **refutada** y se retira: $\rho$
> alcanza su valor final ya en $d/\ell_s=0.1$, mucho antes de $\ell_s/2$. El supuesto de que el
> cambio de $\rho$ lo domina la re-ponderación de la dispersión simple es falso — lo domina la
> excursión de (ii), que es un mecanismo distinto y más interesante.

```python
# ===========================================================================
# MM3a -- (i) el ancho mide l*, (ii) la excursion del apice, (iii) el test de
#         regla, ahora con los DOS observables y veredicto calculado
# ===========================================================================
L_SMALL = min(min(c.l_entry, c.l_back) for c in CAMP.values())
L_LARGE = max(max(c.l_entry, c.l_back) for c in CAMP.values())
LS = list(CAMP.values())[0].ls_common

# --- (i) medios HOMOGENEOS: q_1/2 escala como 1/l*, con l_s identico -------
WIDTH_HOM = {}
for p in pols:
    qs = {}
    for med in ("small", "large"):
        vals = [level_q(x[0], x[1], 0.5) for x in CT.get((p, med), [])]
        vals = [v for v in vals if np.isfinite(v)]
        if vals:
            qs[med] = (float(np.mean(vals)), float(np.std(vals, ddof=1) / np.sqrt(len(vals)))
                       if len(vals) > 1 else 0.0)
    if len(qs) == 2:
        r = qs["large"][0] / qs["small"][0]
        r_pred = L_SMALL / L_LARGE
        WIDTH_HOM[p] = dict(ratio=r, pred=r_pred, dev=100 * (r / r_pred - 1))
        print(f"[MM3 {p:8s}] homogeneos: q_1/2(grande)/q_1/2(pequena) = {r:.4f}"
              f"   predicho l*_peq/l*_gra = {r_pred:.4f}   ({100*(r/r_pred-1):+.1f} %)")
print("      -> con l_s identico en los dos medios, el ancho sigue a l*: la regla")
print("         de transporte queda establecida sobre medios homogeneos, sin bicapa")

# --- (ii) excursion del apice fuera del intervalo de los constituyentes ----
RHO_EXC = {}
for (o, p), D in SCAL.items():
    if p != "linear":
        continue
    r, s = D["rho"], D["s_rho"]
    up = r[-1] > r[0]
    i = int(np.argmax(r)) if up else int(np.argmin(r))
    # limite que acota por ese lado: el mayor (o menor) de los dos extremos
    ends = [0, len(r) - 1]
    j = ends[int(np.argmax([r[0], r[-1]]))] if up else ends[int(np.argmin([r[0], r[-1]]))]
    lim, s_lim = r[j], s[j]
    d = r[i] - lim
    sig = abs(d) / np.hypot(s[i], s_lim)
    RHO_EXC[o] = dict(m_s=D["m_s"][i], rho_ext=r[i], lim=lim,
                      pct=100 * d / lim, sigma=sig, lo=min(r[0], r[-1]), hi=max(r[0], r[-1]))
    print(f"[MM3 {o:7s}] rho: constituyentes {r[0]:.4f} / {r[-1]:.4f}"
          f" | extremo {r[i]:.4f} en d/l_s = {D['m_s'][i]:.2f}"
          f"  ->  excursion {100*d/lim:+.1f} %  ({sig:.1f} sigma)")
if len(RHO_EXC) == 2:
    print("      -> los DOS ordenes excursionan, con signos opuestos y el extremo en el")
    print("         mismo d/l_s. Con mu_s igualado las dos capas de entrada son")
    print("         opticamente identicas a igual z: el mecanismo de 5.3 queda contrastado,")
    print("         no solo reproducido. Un promedio ponderado no puede salir del intervalo.")

# --- (iii) peso de la capa de entrada (lo usan MM4 y MM7) -----------------
WEIGHT = {}
for (o, p), c in CAMP.items():
    if (o, p, "entry") not in CTRL or (o, p, "back") not in CTRL:
        continue
    q_ref, E_entry = CTRL[(o, p, "entry")]
    _,     E_back = CTRL[(o, p, "back")]
    rows = []
    for L in c.layers:
        q, E = PROF[(o, p, L.z)]
        w, rms = fit_two_cones(q_ref, on_ref_grid(q, E, q_ref), E_entry, E_back)
        ws = [fit_two_cones(q_ref,
                            on_ref_grid(*profile_one(c.sweep, kk, c.pol, c.channel,
                                                     c.l_anchor), q_ref),
                            E_entry, E_back)[0] for kk in L.keys]
        rows.append(dict(z=L.z, m_s=L.m_s, m_star=L.m_star, w=w,
                         s_w=t_ci(len(ws)) * np.std(ws, ddof=1) / np.sqrt(len(ws)),
                         rms_two=rms))
    WEIGHT[(o, p)] = pd.DataFrame(rows)

# --- test de regla: profundidad de transicion en las dos normalizaciones ---
def d50(x, y, tail=3):
    """Profundidad a la que el observable ha recorrido la mitad de su rango."""
    y = np.asarray(y, float)
    lvl = 0.5 * (y[0] + np.mean(y[-tail:]))
    for i in range(1, len(y)):
        if (y[i-1] - lvl) * (y[i] - lvl) <= 0 and y[i] != y[i-1]:
            return float(x[i-1] + (lvl - y[i-1]) * (x[i] - x[i-1]) / (y[i] - y[i-1]))
    return np.nan


RULER = {}
for name, getter in (("q_1/2", lambda o: (SCAL[(o, POL_GATE)]["m_s"], SCAL[(o, POL_GATE)]["qh"])),
                     ("w",     lambda o: (WEIGHT[(o, POL_GATE)]["m_s"].values,
                                          WEIGHT[(o, POL_GATE)]["w"].values))):
    try:
        d_ls = {o: d50(*getter(o)) for o in orders}
        d_st = {o: d_ls[o] * LS / CAMP[(o, POL_GATE)].l_entry for o in orders}
        r_ls = d_ls[orders[1]] / d_ls[orders[0]]
        r_st = d_st[orders[1]] / d_st[orders[0]]
        RULER[name] = dict(d_ls=d_ls, r_ls=r_ls, r_st=r_st,
                           win="l_s" if abs(np.log(r_ls)) < abs(np.log(r_st)) else "l*_entry")
        print(f"[MM3 regla | {name:6s}] d50 = "
              + ", ".join(f"{o} {d_ls[o]:.2f} l_s" for o in orders))
        print(f"                    razon {orders[1]}/{orders[0]}: en l_s {r_ls:.3f} "
              f"(|ln| {abs(np.log(r_ls)):.2f}), en l*_ent {r_st:.3f} "
              f"(|ln| {abs(np.log(r_st)):.2f})  ->  gana {RULER[name]['win']}")
    except Exception as err:
        print(f"[MM3 regla | {name}] no evaluable: {err}")

if len(RULER) == 2 and len({v["win"] for v in RULER.values()}) == 2:
    print("\n      VEREDICTO: los dos observables se contradicen -> con contraste "
          f"l*_max/l*_min = {L_LARGE/L_SMALL:.2f} el experimento NO discrimina")
    print("      entre l_s y l*_entrada como regla de profundidad de sondeo.")
    print("      El poder del test ES el contraste; hace falta una familia matched_mus")
    print("      con contraste ~5-10 (p.ej. 35 nm contra 300-500 nm) para decidirlo.")
elif len(RULER) == 2:
    print(f"\n      VEREDICTO: los dos observables coinciden -> gana "
          f"{list(RULER.values())[0]['win']}")
```

## MM4 — Una sola escala contra superposición de los dos controles

> **Corrección respecto de la primera ejecución.** El modelo de "cono único" era AWM, y su RMS sobre
> los propios controles homogéneos valía $1.3\times10^{-2}$. Eso **no es ruido**: es el fallo de AWM
> al describir el cono polarizado medido (la no-universalidad del prefactor de §5.1). Con ese "piso"
> el test no tenía poder (1.7–2.1×) y en un orden el ajuste de dos conos salía peor que el de uno,
> lo cual es absurdo.

Ahora el patrón de "una sola escala" es el **control homogéneo medido con la abscisa estirada**,
$\eta=1+A\,[\eta_{\rm ctrl}(q/s)-1]$, dos parámetros. No supone forma de línea alguna: pregunta
exactamente lo que hay que preguntar — *¿existe un reescalado angular de un medio homogéneo que
reproduzca este perfil?* Con eso el piso pasa a ser estadístico y aparecen dos referencias legítimas:

- **piso estadístico**: describir un control con el *mismo* medio medido en la otra campaña;
- **piso de forma**: describir un constituyente con el *otro*, reescalado. Cota inferior real.

El segundo modelo es la superposición de los dos controles medidos, $\eta_d-1=w[\eta_{\rm ent}-1]+
(1-w)[\eta_{\rm fondo}-1]$, un parámetro por espesor. El residuo que quede mide los caminos híbridos.

El panel (b) contrasta las dos leyes de saturación de $w(m)$: la de cola pesada $m/(m+\zeta)$ con
$\zeta\approx1+\tilde b$ que predice C.7 desde la estadística de excursiones brownianas, y la
exponencial. **Se decide por $\chi^2$, no por lo que dice la teoría**, y la celda imprime el ganador.

```python
# ===========================================================================
# MM4 -- una sola escala contra superposicion, con la forma de linea MEDIDA
#   Cambio respecto de la version anterior: el patron ya no es AWM sino el
#   control homogeneo medido, con la abscisa estirada. Antes el "piso" de
#   1.3e-2 era el fallo de AWM al describir el cono polarizado, no ruido, y
#   por eso el test no tenia poder (1.7-2.1x). Ahora el piso es estadistico.
# ===========================================================================
STRETCH = {}
for (o, p), c in CAMP.items():
    if p != POL_GATE or (o, p, "entry") not in CTRL or (o, p, "back") not in CTRL:
        continue
    q_ref, E_entry = CTRL[(o, p, "entry")]
    _,     E_back = CTRL[(o, p, "back")]

    # piso estadistico: el mismo medio, medido en la otra campania
    floor_stat, med = np.nan, species_of(c, "entry")
    for oo in orders:
        if oo == o or (oo, p) not in CAMP:
            continue
        for which in ("entry", "back"):
            if (oo, p, which) in CTRL and species_of(CAMP[(oo, p)], which) == med:
                twin = CTRL[(oo, p, which)]
                floor_stat = fit_stretched(q_ref, E_entry, twin[0], twin[1])[2]

    # piso de forma: describir un constituyente con el OTRO, reescalado
    floor_shape = fit_stretched(q_ref, E_entry, q_ref, E_back)[2]

    rows = []
    for L in c.layers:
        q, E = PROF[(o, p, L.z)]
        Eg = on_ref_grid(q, E, q_ref)
        best = min(fit_stretched(q_ref, Eg, q_ref, E_entry),
                   fit_stretched(q_ref, Eg, q_ref, E_back), key=lambda r: r[2])
        w, rms2 = fit_two_cones(q_ref, Eg, E_entry, E_back)
        rows.append(dict(z=L.z, m_s=L.m_s, m_star=L.m_star,
                         s_scale=best[1], rms_single=best[2], w=w, rms_two=rms2))
    df = pd.DataFrame(rows)
    df.attrs["floor_stat"] = floor_stat
    df.attrs["floor_shape"] = floor_shape
    STRETCH[o] = df

    print(f"[MM4 {o:7s}] pisos: estadistico {floor_stat:.2e} | forma entre "
          f"constituyentes {floor_shape:.2e}")
    print(f"           una escala (control estirado): RMS max {df['rms_single'].max():.2e}"
          f"   = {df['rms_single'].max()/floor_shape:.1f}x el piso de forma")
    print(f"           superposicion de los dos controles: RMS max {df['rms_two'].max():.2e}"
          f"   = {df['rms_two'].max()/floor_shape:.1f}x")
    imax = int(df["rms_single"].idxmax())
    print(f"           peor espesor para una sola escala: d/l_s = {df.loc[imax,'m_s']:.2f}")
```

```python
# ===========================================================================
# MM4 -- figura y ley de saturacion de w(m)
#   La forma de cola pesada w = m/(m+zeta) (C.7, estadistica de excursiones
#   brownianas) se contrasta contra la exponencial. El ganador se decide por
#   chi2, no por lo que dice la teoria.
# ===========================================================================
def w_heavy(m_, zeta): return m_ / (m_ + zeta)
def w_exp(m_, zeta):   return 1.0 - np.exp(-m_ / zeta)

apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.28})

WFIT = {}
for o, df in STRETCH.items():
    st = STYLE[o]
    ax1.plot(df["m_s"] + 1e-3, df["rms_single"], st["ls"], marker=st["marker"], ms=4,
             lw=0.9, color=st["color"], label=f"{o}: single scale")
    ax1.plot(df["m_s"] + 1e-3, df["rms_two"], st["ls"], marker="^", ms=4, lw=0.9,
             color=st["color"], alpha=0.55, mfc="none", label=f"{o}: superposition")
ax1.axhline(np.nanmean([d.attrs["floor_shape"] for d in STRETCH.values()]),
            color="k", lw=0.7, ls=":", label="constituent shape floor")
ax1.axhline(np.nanmean([d.attrs["floor_stat"] for d in STRETCH.values()]),
            color="k", lw=0.7, ls="--", label="statistical floor")

for o in orders:
    if (o, POL_GATE) not in WEIGHT:
        continue
    df, st = WEIGHT[(o, POL_GATE)], STYLE[o]
    good = np.isfinite(df["w"]) & (df["m_star"] > 1e-3)
    mm, ww = df["m_star"][good].values, df["w"][good].values
    sw = np.maximum(df["s_w"][good].values, 1e-3)
    res = {}
    for name, fn, p0 in (("heavy", w_heavy, ZETA_TEO), ("exp", w_exp, 0.5)):
        try:
            popt, pcov = curve_fit(fn, mm, ww, p0=[p0], sigma=sw,
                                   absolute_sigma=True, maxfev=20000)
            chi2 = float(np.sum(((ww - fn(mm, *popt)) / sw) ** 2))
            res[name] = (float(popt[0]), float(np.sqrt(pcov[0, 0])), chi2, len(mm) - 1)
        except Exception as err:
            print(f"[MM4 {o}] ajuste {name} fallo: {err}")
    WFIT[o] = res
    if {"heavy", "exp"} <= set(res):
        win = "exp" if res["exp"][2] < res["heavy"][2] else "heavy"
        print(f"[MM4 {o:7s}] cola pesada: zeta = {res['heavy'][0]:.3f} +- {res['heavy'][1]:.3f}"
              f"  chi2/ndf = {res['heavy'][2]:.1f}/{res['heavy'][3]}"
              f"   (teoria C.7: {ZETA_TEO:.3f})")
        print(f"           exponencial: escala = {res['exp'][0]:.3f} +- {res['exp'][1]:.3f}"
              f"  chi2/ndf = {res['exp'][2]:.1f}/{res['exp'][3]}")
        msg = ("EXPONENCIAL: la ley de cola pesada C.7 queda rechazada" if win == "exp"
               else "cola pesada, como predice C.7")
        print(f"           -> gana la {msg}")
    ax2.errorbar(df["m_star"] + 1e-3, df["w"], yerr=df["s_w"], fmt=st["marker"], ms=4,
                 lw=0, elinewidth=0.9, color=st["color"], capsize=2, label=st["label"])
    mgrid = np.geomspace(max(df["m_star"].min(), 1e-2), df["m_star"].max(), 200)
    if "exp" in res:
        ax2.plot(mgrid, w_exp(mgrid, res["exp"][0]), st["ls"], lw=0.9, color=st["color"])

mgrid = np.geomspace(1e-2, 12, 200)
ax2.plot(mgrid, w_heavy(mgrid, ZETA_TEO), "k:", lw=0.9,
         label=rf"C.7: $m/(m+{ZETA_TEO:.2f})$")

ax1.set_xscale("log"); ax1.set_yscale("log")
ax1.set_xlabel(r"$d/\ell_s$"); ax1.set_ylabel("fit RMS")
ax1.set_title("(a) single scale vs superposition", loc="left", fontsize="small")
ax1.grid(alpha=0.15, which="both"); ax1.legend(fontsize=5.5, frameon=False)
ax2.set_xscale("log"); ax2.set_ylim(-0.05, 1.2)
ax2.set_xlabel(r"$d/\ell^*_{\rm entry}$"); ax2.set_ylabel(r"entry weight $w$")
ax2.set_title("(b) saturation law", loc="left", fontsize="small")
ax2.grid(alpha=0.15, which="both"); ax2.legend(fontsize=6, frameon=False)
fig.savefig(FIGDIR / "MM4_dos_conos.pdf", bbox_inches="tight")
plt.show()
```

## MM5 — El orden de apilamiento es un observable

Las dos campañas comparten especies, fracciones de volumen y profundidades absolutas de interfaz, y
—éste es el punto de la familia— **$\mu_s$ idéntico en todas partes**: las dos muestras contienen la
misma cantidad de dispersión por unidad de longitud, sólo reordenada en profundidad.

> **Corrección de encuadre.** En $d\to0$ y $d\to\infty$ las dos campañas **no** tienen el mismo
> contenido: `normal` degenera en el medio grande y `inverse` en el pequeño, y viceversa. La
> diferencia en los extremos es simplemente la que separa a los dos homogéneos, y no mide
> no-conmutatividad. El observable es el **exceso** de $|\Delta|$ en los espesores intermedios sobre
> esa línea base — que es lo que la celda reporta ahora. La comparación anterior contra la brecha de
> Jensen de §5.2 se retira: no comparaba lo mismo.

Con esa base, lo que queda es el exceso intermedio y, en el canal circular, un **cambio de signo** de
$\Delta$ con el espesor: el mismo par de medios, reordenado, invierte el sentido del efecto.

```python
# ===========================================================================
# MM5 -- efecto del orden, con la linea base correcta
#   Correccion de encuadre: en d->0 y d->inf las dos campanias NO tienen el
#   mismo contenido (normal->grande, inverse->pequena y viceversa), asi que
#   Delta en los extremos es simplemente la diferencia entre los dos medios
#   homogeneos y no mide no-conmutatividad. El observable es el EXCESO de
#   |Delta| en los espesores intermedios sobre esa linea base de extremos.
# ===========================================================================
apply(width_frac=1.0)
fig, axes = plt.subplots(1, len(pols), figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                         squeeze=False, sharey=True, gridspec_kw={"wspace": 0.12})

ORD_ROWS = []
for ip, p in enumerate(pols):
    ax = axes[0][ip]
    if ("normal", p) not in CAMP or ("inverse", p) not in CAMP:
        ax.axis("off"); continue
    A, B = CAMP[("normal", p)], CAMP[("inverse", p)]
    n_pair = min(len(A.layers), len(B.layers))
    cmap = plt.cm.viridis(np.linspace(0, 0.92, n_pair))
    q_ref = PROF[("normal", p, A.layers[0].z)][0]
    for col, i in zip(cmap, range(n_pair)):
        qa, Ea = PROF[("normal", p, A.layers[i].z)]
        qb, Eb = PROF[("inverse", p, B.layers[i].z)]
        d = on_ref_grid(qa, Ea, q_ref) - on_ref_grid(qb, Eb, q_ref)
        w = q_ref <= CONV["q_view"]
        ax.plot(q_ref[w], d[w], color=col, lw=0.9,
                label=rf"${A.layers[i].m_s:.2f}$" if ip == 0 else None)
        j = int(np.argmax(np.abs(d[w])))
        ORD_ROWS.append(dict(pol=p, z=A.layers[i].z, m_s=A.layers[i].m_s,
                             ext=float(d[w][j]), q_ext=float(q_ref[w][j]),
                             rms=float(np.sqrt(np.mean(d[w] ** 2))),
                             endpoint=(i == 0 or i == n_pair - 1)))
    ax.axhline(0, c="k", lw=0.6)
    ax.set_xlabel(r"$q$")
    ax.set_title(f"({'ab'[ip]}) {p}", loc="left", fontsize="small")
    ax.grid(alpha=0.15)
axes[0][0].set_ylabel(r"$\eta_{\rm normal}-\eta_{\rm inverse}$")
axes[0][0].legend(ncol=3, columnspacing=0.7, handlelength=1.0, fontsize=6,
                  frameon=False, title=r"$d/\ell_s$", title_fontsize=6)
fig.savefig(FIGDIR / "MM5_orden.pdf", bbox_inches="tight")
plt.show()

DF_ORD = pd.DataFrame(ORD_ROWS)
ORD_SUM = {}
if len(DF_ORD):
    display(DF_ORD.round(4))
    for p, sub in DF_ORD.groupby("pol"):
        base = sub[sub["endpoint"]]["ext"].abs().max()      # diferencia de medios puros
        mid = sub[~sub["endpoint"]]
        i = mid["ext"].abs().idxmax()
        ORD_SUM[p] = dict(base=float(base), peak=float(abs(mid.loc[i, "ext"])),
                          ratio=float(abs(mid.loc[i, "ext"]) / base),
                          z=float(mid.loc[i, "z"]), m_s=float(mid.loc[i, "m_s"]))
        print(f"[MM5 {p:8s}] linea base de extremos (medios puros distintos) = {base:.4f}")
        print(f"           |Delta| max en espesores intermedios = {abs(mid.loc[i,'ext']):.4f}"
              f"  en z = {mid.loc[i,'z']:.2f} um (d/l_s = {mid.loc[i,'m_s']:.2f})")
        print(f"           exceso sobre la linea base = {abs(mid.loc[i,'ext'])/base:.2f}x")
    if "circular" in DF_ORD["pol"].values:
        sub = DF_ORD[DF_ORD["pol"] == "circular"].sort_values("m_s")
        sg = np.sign(sub["ext"].values)
        ch = np.where(np.diff(sg) != 0)[0]
        if len(ch):
            i = int(ch[0])
            print(f"[MM5 circular] Delta cambia de signo entre d/l_s = "
                  f"{sub['m_s'].values[i]:.2f} y {sub['m_s'].values[i+1]:.2f}")
    print("      Con mu_s identico en todas partes, la unica diferencia entre las dos")
    print("      muestras es el reordenamiento en profundidad de la MISMA dispersion.")
```

## MM6 — Cono con compuerta temporal

Dentro de un medio homogéneo el cono con compuerta se angosta como $t^{-1/2}$ y, a $t$ fijo, su ancho
escala como $(\ell^*)^{-1/2}$ [Eq. 45]. Con contraste 1.82 el rango dinámico total es $\sqrt{1.82}=1.35$,
así que el cociente crudo $R(t)$ de §5.3 es demasiado poco sensible; se usa la coordenada de
interpolación entre los **dos controles medidos**,
$\lambda(t)=[\ln q_{1/2}^{d}-\ln q_{1/2}^{\rm ent}]/[\ln q_{1/2}^{\rm fondo}-\ln q_{1/2}^{\rm ent}]$,
que normaliza por todo el contraste disponible y cancela a primer orden el piso pre-difusivo común.

> **Corrección respecto de la primera ejecución.** El filtro de alcance era unilateral: excluía
> interfaces demasiado profundas pero no demasiado someras. Para $d/\ell^*_{\rm ent}=0.095$ el $t^*$
> predicho es $0.013\,\tau^*$ y el primer bin temporal está en $0.365\,\tau^*$ — 28 veces más tarde —
> así que el cruce hallado allí era ruido puro y producía un error del 4543 % en la inversión. Ahora
> se exige además $t^*_{\rm pred} > K\cdot t_{\rm primer\ bin}$.

Aun con el filtro, $\alpha$ sale de 2.6 a 8.9 punto a punto contra el $3/2$ teórico. **Es el mismo
efecto de piso pre-difusivo por el que $t^*\propto d^2$ se retiró en §5.3, y aquí tampoco se cita.**
Lo que sí sobrevive sin parámetros es el plateau tardío $R_\infty=\sqrt{\ell^*_{\rm ent}/\ell^*_{\rm fondo}}$,
con su **cambio de signo** entre órdenes — y ése es el resultado temporal, y la base de MM7.

```python
# ===========================================================================
# MM6a -- anchos con compuerta temporal
# ===========================================================================
TIME_STRIDE = 1
POL_TIME = "circular" if any(p == "circular" for (_, p) in CAMP) else "linear"


def noise_tail(q, E, q_min=CONV["q_view"]):
    m = q > q_min
    return float(np.std(E[m])) if np.count_nonzero(m) > 5 else np.inf


def gated_width(c, keys):
    """(t/tau*_ent, q_1/2) por bin temporal, con compuerta de S/N."""
    ts, ws = [], []
    for it in range(1, c.n_time + 1, TIME_STRIDE):
        try:
            q, E = profile_sum(c.sweep, keys, c.pol, c.channel, c.l_anchor,
                               time_index=int(it))
        except Exception:
            continue
        e0 = apex_value(q, E)
        if not np.isfinite(e0) or (e0 - 1.0) < CONV["snr_min_time"] * noise_tail(q, E):
            continue
        h = level_q(q, E, 0.5)
        if np.isfinite(h) and h > 0:
            ts.append((it - 0.5) * c.dt_sim / c.tau_entry)
            ws.append(h)
    return np.asarray(ts), np.asarray(ws)


WID, WCTL = {}, {}
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    for which in ("entry", "back"):
        if c.ctrl[which]:
            WCTL[(o, which)] = gated_width(c, c.ctrl[which])
    for L in c.layers:
        WID[(o, L.z)] = gated_width(c, L.keys)
    print(f"[MM6 {o:7s}] bins validos -- controles: "
          f"entrada {len(WCTL.get((o,'entry'), ([],[]))[0])}, "
          f"fondo {len(WCTL.get((o,'back'), ([],[]))[0])};  "
          f"bicapas: {[len(WID[(o, L.z)][0]) for L in c.layers]}")
```

```python
# ===========================================================================
# MM6b -- lambda(t), t* con filtro de DOS lados, y el plateau tardio
#   Correccion: el filtro de alcance anterior era unilateral (solo excluia
#   interfaces demasiado PROFUNDAS). Una interfaz demasiado somera tiene su t*
#   predicho por debajo del primer bin temporal y el cruce hallado es ruido:
#   asi entraba m*=0.095 con t*=28.9 tau* cuando lo predicho era 0.013 tau*.
#   Ahora se exige ademas t*_pred > K_FIRST veces el primer bin.
# ===========================================================================
K_FIRST = 3.0        # margen exigido sobre el primer bin temporal


def lam_curve(o, z):
    """Coordenada de interpolacion entre los dos controles MEDIDOS."""
    te, we = WCTL.get((o, "entry"), (np.array([]), np.array([])))
    tb, wb = WCTL.get((o, "back"), (np.array([]), np.array([])))
    td, wd = WID.get((o, z), (np.array([]), np.array([])))
    if min(len(te), len(tb), len(td)) < 8:
        return np.array([]), np.array([])
    lo, hi = max(te.min(), tb.min(), td.min()), min(te.max(), tb.max(), td.max())
    if not (hi > lo):
        return np.array([]), np.array([])
    g = np.geomspace(lo, hi, 60)
    le, lb = np.log(np.interp(g, te, we)), np.log(np.interp(g, tb, wb))
    ld = np.log(np.interp(g, td, wd))
    den = lb - le
    with np.errstate(divide="ignore", invalid="ignore"):
        return g, np.where(np.abs(den) > 1e-6, (ld - le) / den, np.nan)


def first_crossing(x, y, level=0.5, persist=2):
    above = y > level
    for i in range(1, len(y) - persist):
        if (not above[i - 1]) and np.all(above[i:i + persist]):
            if y[i] == y[i - 1]:
                return float(x[i])
            return float(x[i - 1] + (level - y[i - 1]) * (x[i] - x[i - 1]) / (y[i] - y[i - 1]))
    return np.nan


apply(width_frac=1.0)
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(TEXTWIDTH_IN, 0.36 * TEXTWIDTH_IN),
                                    gridspec_kw={"wspace": 0.34})

TSTAR = {}
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    st = STYLE[o]
    t_first = 0.5 * c.dt_sim / c.tau_entry
    m_min = np.sqrt(K_FIRST * t_first / ALPHA_TEO)
    print(f"[MM6 {o:7s}] primer bin = {t_first:.3f} tau*  ->  se exige "
          f"d/l*_ent > {m_min:.2f} para que t* sea resoluble")

    for which, lsty in (("entry", "-"), ("back", "--")):
        if (o, which) in WCTL:
            t_, w_ = WCTL[(o, which)]
            ax1.plot(t_, w_, lsty, color=st["color"], lw=1.4, alpha=0.9,
                     label=f"{o}: {which} control")
    cmap = plt.cm.viridis(np.linspace(0, 0.92, len(c.layers)))
    rows = []
    for col, L in zip(cmap, c.layers):
        t_, w_ = WID[(o, L.z)]
        if len(t_):
            ax1.plot(t_, w_, "-", color=col, lw=0.6, alpha=0.75)
        g, lam = lam_curve(o, L.z)
        if not len(g):
            continue
        ax2.plot(g, lam, "-", color=col, lw=0.7, alpha=0.85)
        usable = bool(L.reachable) and (L.m_star > m_min)
        ts = first_crossing(g, lam, 0.5) if usable else np.nan
        rows.append(dict(z=L.z, m_s=L.m_s, m_star=L.m_star, t_star=ts, usable=usable,
                         lam_late=float(np.nanmedian(lam[-max(3, len(lam)//6):])),
                         alpha_pt=ts / L.m_star ** 2 if np.isfinite(ts) else np.nan))
    TSTAR[o] = pd.DataFrame(rows)

    df = TSTAR[o].dropna(subset=["t_star"])
    if len(df) >= 2:
        u, t = df["m_star"].values ** 2, df["t_star"].values
        a_fit = float(np.sum(u * t) / np.sum(u ** 2))
        TSTAR[o].attrs["alpha"] = a_fit
        TSTAR[o].attrs["alpha_spread"] = float(df["alpha_pt"].max() / df["alpha_pt"].min())
        ax3.plot(u, t, st["marker"], ms=5, color=st["color"],
                 label=rf"{o}, $\alpha={a_fit:.2f}$")
        print(f"           alpha por punto: {np.round(df['alpha_pt'].values, 2)}"
              f"   (dispersion {df['alpha_pt'].max()/df['alpha_pt'].min():.1f}x)")
        print(f"           alpha global = {a_fit:.2f} contra 3/2 teorico -> "
              f"{'INCONSISTENTE' if abs(np.log(a_fit/ALPHA_TEO)) > 0.4 else 'compatible'}")
    else:
        print(f"           menos de 2 espesores utilizables: t* no evaluable")

ax1.set_xscale("log"); ax1.set_yscale("log")
ax1.set_xlabel(r"$t/\tau^*_{\rm entry}$"); ax1.set_ylabel(r"$q_{1/2}(t)$")
ax1.set_title("(a) gated cone width", loc="left", fontsize="small")
ax1.grid(alpha=0.15, which="both"); ax1.legend(fontsize=5.5, frameon=False)
for lv in (0, 0.5, 1):
    ax2.axhline(lv, c="k", lw=0.6, ls=":" if lv == 0.5 else "-", alpha=0.7)
ax2.set_xscale("log"); ax2.set_ylim(-0.3, 1.4)
ax2.set_xlabel(r"$t/\tau^*_{\rm entry}$"); ax2.set_ylabel(r"$\lambda(t)$")
ax2.set_title("(b) interpolation coordinate", loc="left", fontsize="small")
ax2.grid(alpha=0.15, which="both")
uu = np.geomspace(1e-2, 1e2, 50)
ax3.plot(uu, ALPHA_TEO * uu, "k-", lw=0.8, alpha=0.7, label=r"$\alpha=3/2$")
ax3.set_xscale("log"); ax3.set_yscale("log")
ax3.set_xlabel(r"$(d/\ell^*_{\rm entry})^2$"); ax3.set_ylabel(r"$t^*/\tau^*_{\rm entry}$")
ax3.set_title("(c) break time", loc="left", fontsize="small")
ax3.grid(alpha=0.15, which="both"); ax3.legend(fontsize=6, frameon=False)
fig.savefig(FIGDIR / "MM6_tiempo.pdf", bbox_inches="tight")
plt.show()

# --- lo que SI es sin parametros: el plateau tardio ------------------------
print("\n[MM6] plateau tardio de anchos entre los dos controles (sin parametros):")
RINF = {}
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    te, we = WCTL.get((o, "entry"), (np.array([]), np.array([])))
    tb, wb = WCTL.get((o, "back"), (np.array([]), np.array([])))
    if len(te) > 5 and len(tb) > 5:
        g = np.geomspace(max(te.min(), tb.min()), min(te.max(), tb.max()), 40)[-10:]
        R = float(np.median(np.interp(g, tb, wb) / np.interp(g, te, we)))
        RINF[o] = dict(meas=R, pred=c.R_inf, dev=100 * (R / c.R_inf - 1))
        print(f"      {o:8s}: medido {R:.3f}   predicho sqrt(l*_ent/l*_fondo) = {c.R_inf:.3f}"
              f"   ({100*(R/c.R_inf-1):+.1f} %)   el cono "
              f"{'se ensancha' if c.R_inf > 1 else 'se estrecha'}")
print("      El cambio de signo entre ordenes se cumple; es el resultado temporal solido.")
print("      El escalamiento t* ~ d^2, en cambio, sigue contaminado por efectos de piso")
print("      pre-difusivos (misma razon por la que se retiro en 5.3): no se cita.")
```

## MM7 — Inversión: espesor y medio enterrado

> **Corrección respecto de la primera ejecución.** Se **retira la ruta $t^*$** para el espesor: daba
> errores del 80 % al 4500 % porque hereda íntegros los efectos de piso pre-difusivos de MM6. Queda
> una sola ruta, $w(d)$, que es monótona, tiene barras de error por réplica y —lo importante— se
> valida **dejando fuera el punto que se invierte**: la curva $w(m)$ se calibra sin él y luego se
> invierte, de modo que el error citado no está inflado por ajustar al propio dato.

**(1) Espesor por $w(d)$.** Con los dos constituyentes caracterizados, un solo perfil CBS estacionario
da el espesor de la capa. El rango útil es donde $w$ no ha saturado; fuera de él la barra de error
diverge, y ése es el límite honesto del método, no un fallo.

**(2) El medio enterrado.** Del plateau tardío, $\ell^*_{\rm fondo}=\ell^*_{\rm ent}/R_\infty^2$; y
como $\mu_s$ está igualado y $\ell_s$ es conocido, $g_{\rm fondo}=1-\ell_s/\ell^*_{\rm fondo}$, y de
ahí el radio por Mie. **Esta cadena es privativa de la familia $\mu_s$ igualado**: sin ese invariante,
$\ell^*$ por sí solo no separa densidad de tamaño.

La precisión en el radio es muy asimétrica entre los dos medios, y no por ruido: es el jacobiano
$dg/da$. Cerca de Rayleigh $g$ es pequeño y casi plano, así que un error de pocos por ciento en
$\ell^*$ se amplifica a decenas por ciento en $g$ y más aún en $a$. Es una afirmación de
**condicionamiento del problema inverso**, cuantificable y citable por sí sola: el método dimensiona
bien dispersores anisótropos y mal los casi isótropos.

```python
# ===========================================================================
# MM7 -- inversion
#   Cambio respecto de la version anterior: se RETIRA la ruta t*, que daba
#   errores del 80-4500 % porque hereda los efectos de piso pre-difusivos de
#   MM6. Queda una sola ruta de espesor, w(d), que es monotona, tiene barras
#   de error y se valida dejando fuera el punto que se invierte.
# ===========================================================================
ls_c = list(CAMP.values())[0].ls_common


def invert_w(df, i, model=w_exp):
    """Leave-one-out: se calibra w(m) SIN el punto i y luego se invierte ese
    punto. Evita el optimismo de invertir con una curva ajustada al propio
    dato."""
    good = np.isfinite(df["w"]) & (df["m_s"] > 1e-3)
    good.iloc[i] = False
    mm, ww = df["m_s"][good].values, df["w"][good].values
    sw = np.maximum(df["s_w"][good].values, 1e-3)
    popt, pcov = curve_fit(model, mm, ww, p0=[0.5], sigma=sw,
                           absolute_sigma=True, maxfev=20000)
    lam, slam = float(popt[0]), float(np.sqrt(pcov[0, 0]))
    w_i, sw_i = float(df["w"].iloc[i]), float(max(df["s_w"].iloc[i], 1e-3))
    if not (0.02 < w_i < 0.96):
        return np.nan, np.nan
    m_hat = -lam * np.log(1.0 - w_i)
    dm_dw = lam / (1.0 - w_i)
    s_m = np.hypot(dm_dw * sw_i, (m_hat / lam) * slam)
    return m_hat * ls_c, s_m * ls_c


INV = []
for o in orders:
    key = (o, POL_GATE)
    if key not in WEIGHT:
        continue
    df = WEIGHT[key].reset_index(drop=True)
    for i in range(len(df)):
        d_hat, s_d = invert_w(df, i)
        if np.isfinite(d_hat):
            INV.append(dict(order=o, z_true=float(df["z"].iloc[i]),
                            d_hat=d_hat, s_d=s_d))
DF_INV = pd.DataFrame(INV)

apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.30})

if len(DF_INV):
    DF_INV["rel_err"] = DF_INV["d_hat"] / DF_INV["z_true"] - 1.0
    zz = np.geomspace(DF_INV["z_true"].min(), DF_INV["z_true"].max(), 50)
    ax1.plot(zz, zz, "k-", lw=0.8, alpha=0.7, label="identity")
    ax1.fill_between(zz, 0.8 * zz, 1.2 * zz, color="0.85", zorder=0, label=r"$\pm20\%$")
    for o, sub in DF_INV.groupby("order"):
        st = STYLE[o]
        ax1.errorbar(sub["z_true"], sub["d_hat"], yerr=sub["s_d"], fmt=st["marker"],
                     ms=4.5, lw=0, elinewidth=0.9, capsize=2, color=st["color"],
                     label=f"{o}, from $w(d)$")
        print(f"[MM7 {o:7s}] espesor por w(d), leave-one-out: mediana "
              f"{100*sub['rel_err'].median():+.1f} %,  |err| max "
              f"{100*sub['rel_err'].abs().max():.1f} %"
              f"   ({len(sub)} puntos, d in [{sub['z_true'].min():.1f}, "
              f"{sub['z_true'].max():.1f}] um)")
    ax1.set_xscale("log"); ax1.set_yscale("log")
    ax1.set_xlabel(r"true $d$ [$\mu$m]"); ax1.set_ylabel(r"recovered $\hat d$ [$\mu$m]")
    ax1.set_title("(a) blind thickness from $w(d)$", loc="left", fontsize="small")
    ax1.grid(alpha=0.15, which="both"); ax1.legend(fontsize=6, frameon=False)
    print("      Rango util: donde w no ha saturado. Fuera de el la barra de error diverge,")
    print("      y ese es el limite honesto del metodo, no un fallo.")

# --- medio enterrado: R_inf -> l*_fondo -> g_fondo ------------------------
BURIED = []
for o in orders:
    if o not in RINF:
        continue
    c = CAMP[(o, POL_TIME)]
    l_hat = c.l_entry / RINF[o]["meas"] ** 2
    BURIED.append(dict(order=o, R_meas=RINF[o]["meas"], R_pred=c.R_inf,
                       l_back_hat=l_hat, l_back_true=c.l_back,
                       g_back_hat=1.0 - ls_c / l_hat,
                       g_back_true=1.0 - ls_c / c.l_back))
DF_BUR = pd.DataFrame(BURIED)
if len(DF_BUR):
    display(DF_BUR.round(4))
    idx = np.arange(len(DF_BUR))
    ax2.bar(idx - 0.18, DF_BUR["g_back_hat"], 0.34, color=COL[2],
            label=r"recovered $g_{\rm back}$")
    ax2.bar(idx + 0.18, DF_BUR["g_back_true"], 0.34, color="0.75",
            label=r"true $g_{\rm back}$")
    ax2.set_xticks(idx); ax2.set_xticklabels(DF_BUR["order"])
    ax2.set_ylabel(r"$g$ of the buried medium")
    ax2.set_title("(b) buried medium recovery", loc="left", fontsize="small")
    ax2.grid(alpha=0.15, axis="y"); ax2.legend(fontsize=6, frameon=False)
    for _, r in DF_BUR.iterrows():
        print(f"[MM7 {r['order']:7s}] l*_fondo {r['l_back_hat']:.1f} vs {r['l_back_true']:.1f} um "
              f"({100*(r['l_back_hat']/r['l_back_true']-1):+.1f} %)   |   "
              f"g_fondo {r['g_back_hat']:.4f} vs {r['g_back_true']:.4f} "
              f"({100*(r['g_back_hat']/r['g_back_true']-1):+.1f} %)")
    print("      La amplificacion del error al pasar de l* a g es el jacobiano dg/da:")
    print("      cerca de Rayleigh g es pequeno y casi plano, asi que un pocos-% en l*")
    print("      se convierte en decenas de % en g. Es condicionamiento del problema")
    print("      inverso, no ruido: el metodo dimensiona bien dispersores anisotropos.")

fig.savefig(FIGDIR / "MM7_inversion.pdf", bbox_inches="tight")
plt.show()
```

### Por qué esta familia y no la de $\phi$ fija

El argumento no es estético. Con $\phi$ fija cambian $\ell_s$ y $\ell^*$ a la vez, así que cualquier
diferencia entre órdenes es atribuible a dos causas y no hay forma de separarlas dentro de esa
campaña. Con $\mu_s$ igualado el contraste es **uno solo** — la anisotropía — y cada efecto tiene un
único responsable. Eso es lo que permite que MM3-i y MM3-ii sean afirmaciones limpias.

El precio está en MM2/MM4: con contraste $\ell^*$ de 1.82 el no-colapso es un factor de pocas
unidades sobre el piso, no las 12–14× de la campaña de alto contraste. Si el capítulo va a apoyarse
sólo en esta familia, **el peso de la afirmación "no hay medio efectivo" debe recaer en la excursión
de $\rho$** (que es de signo, no de magnitud, y por tanto insensible al contraste) y no en el colapso.
La alternativa es conservar la campaña de $\phi$ fija exclusivamente para esa afirmación y usar ésta
para el mecanismo y la metrología; las dos secciones se leen en paralelo porque comparten
observables y protocolo estadístico.

### Verificaciones antes de citar cualquier número

1. **Canales.** La resolución por física elige el canal circular cuyo ápice está más cerca de 2. En
   esta campaña resultó ser el etiquetado `cross`, no `co`. Es coherente con la convención de
   helicidad en retrodispersión (el sentido de giro en el marco del laboratorio se invierte al
   invertirse la dirección de propagación), pero **hay que documentarlo en el diccionario único de
   canales**, porque leer la etiqueta al pie de la letra da el canal equivocado.
2. **Grillas.** Los cuatro `grid_fingerprint` salen distintos. La causa es que $\ell^*$ difiere
   $\sim0.1\%$ entre campañas que sólo cambian la polarización y la semilla, lo que sugiere que
   `derived_quantities` no es determinista ($g$ probablemente por muestreo). Todas las comparaciones
   entre campañas interpolan (`on_ref_grid`), así que el análisis es correcto — pero la afirmación de
   diseño "grillas idénticas bin a bin dentro de la familia" es falsa y afecta a lo que el Cap. 4
   dice sobre reproducibilidad.
3. **Script archivado.** Los datos tienen 10 espesores y $\sim300$ bins angulares; el script suelto
   lista 12 y `N_THETA_1 = 400`. Usa el snapshot que archivó `SweepManager`, no el fichero de
   trabajo.
4. **Espesores que faltan.** $\rho$ ya está saturada en $d/\ell_s=0.1$, de modo que la **subida** de
   la excursión no está resuelta. Re-correr $d/\ell_s = 0.02$ y $0.05$ (dos espesores, 5 réplicas)
   es lo más barato con más rendimiento de toda esta lista: convierte MM3(a) en una curva y no en
   tres puntos.
5. **Sistemático de reciprocidad.** $\eta_+(0)$ vale 1.985 y 1.992, no 2, y el déficit depende del
   tamaño (mayor para la partícula casi Rayleigh). La reciprocidad lo prohíbe para esferas, así que
   es sistemático del motor — consistente con el depósito inconsistente de dispersión simple ya
   documentado. Decláralo como piso sistemático de $\sim1\%$ y no lo absorbas en las barras de error.

# §5.4 — Bicapa estratificada a $\mu_s$ igualado: dos reglas de profundidad en un mismo cono

**Qué distingue a esta campaña.** Las cuatro corridas (`NORMAL`/`INVERSE` $\times$ `PCIR`/`PLIN`)
pertenecen a la familia `matched_mus`: la fracción de volumen de la especie grande se deriva de
$\phi_{\rm L}=4a_{\rm L}\mu_s^{\rm ref}/(3Q_{\rm L})$ de modo que **$\mu_s$ es idéntico en las dos
capas**. Sin absorción, esto implica

$$\ell_s^{\rm sup}=\ell_s^{\rm inf}\equiv\ell_s \quad\text{(común a las dos capas y a los dos órdenes)},
\qquad \frac{\ell^*_{\rm L}}{\ell^*_{\rm S}}=\frac{1-g_{\rm S}}{1-g_{\rm L}}\simeq 1.8 .$$

El único contraste es la **anisotropía**. Eso es lo que hace a esta tanda distinta de §5.3 (donde
$\phi$ era fija y $\ell_s$, $\ell^*$ cambiaban a la vez, con contraste 5.09): allí no se podía separar
cuál de las dos longitudes gobernaba cada observable. Aquí sí, porque a **igual $z$ absoluto**

| | $d/\ell_s$ | $d/\ell^*_{\rm ent}$ |
|---|---|---|
| `normal` (entra la pequeña) | $z/\ell_s$ | $z/\ell^*_{\rm S}$ |
| `inverse` (entra la grande) | $z/\ell_s$ **— el mismo número** | $z/\ell^*_{\rm L}$ **— difiere en 1.8** |

Un observable que colapse contra una de las dos columnas y no contra la otra identifica su longitud
de sondeo **sin ningún parámetro ajustado**. Ese es el eje del notebook.

---

## La tesis del capítulo, en una línea

El cono de la bicapa lleva **dos reglas de profundidad simultáneas y de escalas muy distintas**:

- **Regla corta — el ápice.** $\rho=2-\eta(0)$ es un observable de *un solo vértice* (§1.3 del
  documento de fundamentos). Su peso de dispersión simple es, con $\ell_s$ común,
  $$S(d)\;\propto\;\tfrac12\big[p_{\rm sup}(\pi)\,(1-e^{-2d/\ell_s})+p_{\rm inf}(\pi)\,e^{-2d/\ell_s}\big],$$
  de modo que $X\equiv\rho/(1-\rho)=S/M$ progresa como $1-e^{-2d/\ell_s}$: **escala $\ell_s/2$, idéntica
  en los dos órdenes, con el signo del cambio invertido** (porque $p_{\rm sup}$ y $p_{\rm inf}$ se
  intercambian). Cero parámetros libres.
- **Regla larga — el ancho y la dinámica.** El ancho angular y el cono con compuerta temporal son
  observables de transporte: colapsan contra $d/\ell^*_{\rm ent}$, y el quiebre temporal obedece
  $t^*=\tfrac32\,(d/\ell^*_{\rm ent})^2\,\tau^*_{\rm ent}$ con plateau
  $R_\infty=\sqrt{\ell^*_{\rm ent}/\ell^*_{\rm fondo}}$, que **cambia de signo** al invertir el orden.

De ahí sale la parte metrológica: dos rangos de profundidad complementarios ($d\lesssim2\ell_s$ por el
ápice, $d\sim\ell^*$ por el tiempo) y la inversión del medio enterrado vía $R_\infty\to\ell^*_{\rm fondo}
\to g_{\rm fondo}\to a$ (posible sólo porque $\mu_s$ está igualado y $\ell_s$ es conocido).

- **Constantes teóricas** (cero parámetros): $\tilde b=0.7104$ (Milne), $\zeta_{\rm teo}=1+\tilde b$,

## 1 — Utilidades de reducción

Tres cambios respecto del pipeline de §5.3, todos motivados por problemas concretos observados allí:

1. **Resolución de canales por física, no por etiqueta.** El nombre del canal difiere entre bases
   `linear` y `circular` y ha habido intercambios de etiqueta. En vez de fiarse del nombre, el canal
   que preserva la helicidad se identifica como **aquel cuyo ápice está más cerca de 2 en los
   controles homogéneos** (Prop. 1.1: $\eta_+(0)=2$ protegido por reciprocidad), y el canal
   co-polarizado lineal como el de mayor ápice. La elección se imprime para inspección y se puede
   forzar con `CHANNEL_OVERRIDE`.
2. **Envolvente monótona antes de invertir niveles.** $q_x$ se obtiene sobre
   $\min$-acumulado del exceso normalizado. Un cono es monótonamente decreciente; imponerlo elimina
   los cruces espurios por ruido sin tocar la señal. Se aplica igual a datos, controles y familia
   sintética, así que cualquier sesgo residual se cancela en razones y progresos.
3. **Todo en $q$**, no en mrad. Como el ancla angular es común a las cuatro campañas, $q$ es
   literalmente el mismo índice de bin y las restas entre campañas no requieren interpolación.

```python
# ===========================================================================
# 1a -- resolucion de canales, stitching, linea base
# ===========================================================================
CHANNEL_OVERRIDE = {}      # p.ej. {"circular": "same", "linear": "co"} para forzar


def _basis_for(pol):
    return circular if pol == "circular" else linear


def _chan_names(prof):
    c = prof.coherent
    try:
        return list(c.keys())
    except AttributeError:
        return [a for a in dir(c) if not a.startswith("_")]


def _chan_get(prof, name, which="coherent"):
    obj = getattr(prof, which)
    try:
        return np.asarray(obj[name], float)
    except (TypeError, KeyError):
        return np.asarray(getattr(obj, name), float)


def raw_profiles(sweep, key, pol, channel, time_index=0):
    """(p1, p2) crudos de las dos ventanas angulares, un solo corte azimutal."""
    sw = sweep[key]
    b = _basis_for(pol)
    p1 = cbs_profiles(sw.processed_cbs("farfield_cbs_1"), basis=b,
                      time_index=time_index, reduce=phi_cut(CONV["phi_main"]))
    p2 = cbs_profiles(sw.processed_cbs("farfield_cbs_2"), basis=b,
                      time_index=time_index, reduce=phi_cut(CONV["phi_main"]))
    return p1, p2


def stitch(th1, y1, th2, y2, key="", warn=0.05):
    """Une ventana fina y cola; avisa si el solape no concuerda."""
    th_1 = th1[-1]
    ov = (th2 >= 0.9 * th_1) & (th2 <= th_1)
    if ov.any():
        resid = np.abs(y2[ov] - np.interp(th2[ov], th1, y1))
        if np.max(resid) > warn:
            print(f"[stitch WARN] {key}: max desacuerdo en solape = {np.max(resid):.3f}")
    keep2 = th2 > th_1
    return (np.concatenate([th1, th2[keep2]]),
            np.concatenate([y1, y2[keep2]]))


def tail_baseline(q, y, q_min=CONV["q_min_baseline"], n_min=5, frac=0.85):
    """Mediana de la cola en q comun. Si la ventana no alcanza, usa el 15 % final."""
    q, y = np.asarray(q, float), np.asarray(y, float)
    m = q > q_min
    if np.count_nonzero(m) >= n_min:
        return float(np.median(y[m]))
    m = q >= frac * np.nanmax(q)
    if np.count_nonzero(m) < n_min:
        return float(np.median(y[np.argsort(q)[-n_min:]]))
    return float(np.median(y[m]))
```

```python
# ===========================================================================
# 1b -- perfiles: una replica y sumado sobre replicas
# ===========================================================================
def profile_one(sweep, key, pol, channel, l_anchor, time_index=0):
    """(q, eta) de UNA replica; linea base llevada a 1.
    Se usa SOLO para barras de error de escalares (la unidad estadistica
    independiente es la replica, no el bin)."""
    p1, p2 = raw_profiles(sweep, key, pol, channel, time_index)
    E1 = (_chan_get(p1, channel, "coherent") + eps) / (_chan_get(p1, channel, "incoherent") + eps)
    E2 = (_chan_get(p2, channel, "coherent") + eps) / (_chan_get(p2, channel, "incoherent") + eps)
    th, E = stitch(np.asarray(p1.theta), E1, np.asarray(p2.theta), E2, key)
    q = k * l_anchor * th
    return q, E - tail_baseline(q, E) + 1.0


def profile_sum(sweep, keys, pol, channel, l_anchor, time_index=0):
    """(q, eta) por cociente de intensidades SUMADAS sobre replicas.
    Estimador correcto del valor central: sum(C)/sum(I), no el promedio de
    cocientes por replica (sesgo de estimador de razon, ec. 17-18)."""
    Ic = Ii = th1 = th2 = None
    for key in keys:
        p1, p2 = raw_profiles(sweep, key, pol, channel, time_index)
        c1, c2 = _chan_get(p1, channel), _chan_get(p2, channel)
        b1, b2 = _chan_get(p1, channel, "incoherent"), _chan_get(p2, channel, "incoherent")
        if Ic is None:
            Ic = [np.zeros_like(c1), np.zeros_like(c2)]
            Ii = [np.zeros_like(b1), np.zeros_like(b2)]
            th1, th2 = np.asarray(p1.theta), np.asarray(p2.theta)
        Ic[0] += c1; Ic[1] += c2; Ii[0] += b1; Ii[1] += b2
    keep2 = th2 > th1[-1]
    th = np.concatenate([th1, th2[keep2]])
    Ct = np.concatenate([Ic[0], Ic[1][keep2]])
    It = np.concatenate([Ii[0], Ii[1][keep2]])
    E = (Ct + eps) / (It + eps)
    q = k * l_anchor * th
    return q, E - tail_baseline(q, E) + 1.0
```

```python
# ===========================================================================
# 1c -- escalares robustos a un perfil de dos escalas
# ===========================================================================
def apex_value(q, E, n=None, mode=None):
    """Valor del apice eta(0).

    El pico CBS es NO analitico en theta=0: eta ~ eta(0) - beta*q + O(q^2)
    (cuspide triangular, ec. 12). Promediar los primeros bins subestima el
    apice, y lo subestima MAS para conos estrechos -> mete una dependencia
    espuria con l* en toda cantidad normalizada por eta(0). Extrapolar la recta
    de la cuspide a q=0 elimina ese sesgo a primer orden para CUALQUIER forma
    de linea, y promedia ruido sobre n bins. La celda de calibracion de abajo
    fija n comparando sesgo (piso sintetico) contra ruido (dispersion entre
    replicas)."""
    n = CONV["n_apex"] if n is None else n
    mode = CONV["apex_mode"] if mode is None else mode
    E = np.asarray(E, float)
    if mode == "median":
        return float(np.median(E[:n]))
    p = np.polyfit(np.asarray(q, float)[:n], E[:n], 1)
    return float(p[1])


def excess_norm(q, E, monotone=None):
    """(eta-1)/(eta(0)-1), con envolvente monotona decreciente opcional.
    Un cono es monotonamente decreciente; imponerlo elimina cruces espurios por
    ruido sin tocar la senal, y se aplica igual a datos, controles y familia
    sintetica, asi que cualquier sesgo residual se cancela en razones."""
    monotone = CONV["monotone_levels"] if monotone is None else monotone
    E0 = apex_value(q, E)
    if not np.isfinite(E0) or E0 <= 1.0:
        return None
    y = (np.asarray(E, float) - 1.0) / (E0 - 1.0)
    return np.minimum.accumulate(y) if monotone else y


def level_q(q, E, x):
    """q donde el exceso normalizado cae a la fraccion x del apice."""
    y = excess_norm(q, E)
    if y is None:
        return np.nan
    b = np.where(y < x)[0]
    b = b[b > 0]
    if len(b) == 0:
        return np.nan
    i = int(b[0])
    if y[i] == y[i - 1]:
        return np.nan
    return float(q[i - 1] + (x - y[i - 1]) * (q[i] - q[i - 1]) / (y[i] - y[i - 1]))


def shape_invariant(q, E):
    """S = q_0.25 / q_0.75. Razon de dos escalas del MISMO perfil, luego es
    independiente de la escala: cualquier medio con un unico l* da el mismo S.
    Leerlo como 'dos reglas del mismo cono que en un medio efectivo coinciden'."""
    a, b = level_q(q, E, CONV["x_lo"]), level_q(q, E, CONV["x_hi"])
    return float(a / b) if (np.isfinite(a) and np.isfinite(b) and b > 0) else np.nan


def scalars_one(sweep, key, pol, channel, l_anchor):
    q, E = profile_one(sweep, key, pol, channel, l_anchor)
    e0 = apex_value(q, E)
    return [e0, 2.0 - e0, level_q(q, E, 0.5), shape_invariant(q, E)]


def scalars_stats(sweep, keys, pol, channel, l_anchor):
    """Media y semi-intervalo t al 95 % sobre replicas de (eta0, rho, q_1/2, S).
    La unidad estadistica independiente es la REPLICA, nunca el bin: los bins
    angulares comparten los mismos caminos simulados y estan correlacionados."""
    A = np.asarray([scalars_one(sweep, kk, pol, channel, l_anchor) for kk in keys], float)
    n = len(A)
    return A.mean(0), t_ci(n) * A.std(0, ddof=1) / np.sqrt(n)


# --- forma normalizada y dispersion del colapso (test de auto-similitud) ----
U_GRID = np.linspace(0.15, 4.0, 200)


def normalized_shape(q, E, u_grid=U_GRID):
    """f(u) = [eta-1]/[eta(0)-1] contra u = q/q_(1/2). Elimina amplitud y escala:
    toda familia de UNA sola escala colapsa exactamente, sea cual sea la forma."""
    y = excess_norm(q, E)
    q_h = level_q(q, E, 0.5)
    if y is None or not np.isfinite(q_h) or q_h <= 0:
        return np.full_like(u_grid, np.nan)
    return np.interp(u_grid, np.asarray(q, float) / q_h, y)


def collapse_spread(curves):
    Cm = np.vstack(curves)
    return float(np.sqrt(np.nanmean((Cm - np.nanmean(Cm, axis=0)) ** 2)))


# --- forma de linea AWM (piso sintetico y ajuste de cono unico) -------------
def cone_shape(qq, b=B_MILNE):
    """Cono difusivo AWM normalizado, C(0)=1 (ec. 10 del documento base)."""
    qq = np.maximum(np.asarray(qq, float), 1e-12)
    return (1.0 / (1.0 + 2.0 * b)) / (1.0 + qq) ** 2 * \
           (1.0 + (1.0 - np.exp(-2.0 * b * qq)) / qq)
```

## 2 — Carga de las cuatro campañas

Cada campaña trae, además de los 12 espesores $\times$ 5 réplicas, **dos controles semi-infinitos
explícitos** (`single_layer_small`, `single_layer_large`) con la misma grilla y las mismas réplicas.
Es una mejora sobre §5.3, donde los controles eran los extremos del propio barrido: ahora el límite
homogéneo es una medición independiente, no una extrapolación.

Las clasificaciones se leen de `params_flat` (`layer_kind`, `single_layer_species`, `order`), y el
orden se **re-deriva** de los radios como verificación cruzada. El `grid_fingerprint` debe coincidir
en las cuatro campañas: es la licencia para restar perfiles bin a bin.

```python
# ===========================================================================
# 2a -- estructuras y lector de campania
# ===========================================================================
@dataclass
class Layer:
    z: float                      # espesor absoluto de la capa de entrada [um]
    m_star: float                 # d / l*_entrada   <- regla LARGA
    m_s: float                    # d / l_s          <- regla CORTA (comun a ordenes)
    reachable: bool
    keys: list = field(default_factory=list)


@dataclass
class Campaign:
    tag: str; order: str; pol: str
    sweep: object
    r_entry: float; r_back: float
    l_entry: float; l_back: float          # l* de entrada y de fondo
    ls_common: float                       # l_s (identico en ambas capas)
    g_entry: float; g_back: float
    l_anchor: float                        # ancla angular comun a las 4 campanias
    fingerprint: str
    layers: list
    ctrl: dict                             # 'entry' / 'back' -> lista de keys
    dt_sim: float; t_max_sim: float; n_time: int
    channel: str = ""

    @property
    def m_star_arr(self): return np.array([L.m_star for L in self.layers])
    @property
    def m_s_arr(self):    return np.array([L.m_s for L in self.layers])
    @property
    def z_arr(self):      return np.array([L.z for L in self.layers])
    @property
    def tau_entry(self):
        """tau*_entrada en unidades de simulacion (camino geometrico en um)."""
        return self.l_entry
    @property
    def R_inf(self):
        """Plateau tardio del cociente de anchos: sqrt(l*_ent / l*_fondo)."""
        return float(np.sqrt(self.l_entry / self.l_back))


def _get(p, names, default=np.nan):
    """Lee el primer nombre presente; tolera aplanado por punto o guion bajo."""
    for nm in names:
        for cand in (nm, nm.replace(".", "_"), nm.replace("_", ".")):
            if cand in p:
                try:
                    return float(p[cand])
                except (TypeError, ValueError):
                    return p[cand]
    return default


def _get_str(p, names, default=""):
    v = _get(p, names, default)
    return v if isinstance(v, str) else default


def _find_g(p, which):
    """g de la capa 'top'/'bot': prueba nombres directos y luego busca en las
    claves aplanadas de dq_top / dq_bot."""
    v = _get(p, [f"dq_{which}.anisotropy_g", f"g_{which}", f"anisotropy_g_{which}"])
    if isinstance(v, float) and np.isfinite(v):
        return v
    for kk in p:
        if which in kk and "anisotropy" in kk:
            try:
                return float(p[kk])
            except (TypeError, ValueError):
                pass
    return np.nan
```

```python
# ===========================================================================
# 2b -- carga
# ===========================================================================
def load_campaign(order, pol, folder):
    sw = load_sweep(folder, base_path=Path(BASE_PATH))

    by_z, ctrl = {}, {"small": [], "large": []}
    p_any = None
    for key in sw.keys():
        p = sw[key].params_flat
        p_any = p_any or p
        kind = _get_str(p, ["layer_kind"])
        if kind == "stratified_one_layer" or "single_layer" in key:
            spec = _get_str(p, ["single_layer_species"])
            if not spec:
                spec = "small" if "small" in key else "large"
            ctrl[spec].append(key)
            continue
        z = _get(p, ["z_interface"])
        if not (isinstance(z, float) and np.isfinite(z)):
            mm = re.search(r"z_interface_([0-9.]+)__rep(\d+)", key)
            if not mm:
                print(f"[{order}/{pol}] [skip] {key}")
                continue
            z = float(mm.group(1))
        by_z.setdefault(round(z, 6), []).append(key)

    p0 = sw[sorted(by_z[sorted(by_z)[0]])[0]].params_flat
    r_top, r_bot = _get(p0, ["radius_top"]), _get(p0, ["radius_bot"])
    l_top, l_bot = _get(p0, ["l_star_top"]), _get(p0, ["l_star_bot"])
    ls_top, ls_bot = _get(p0, ["l_s_top"]), _get(p0, ["l_s_bot"])
    g_top, g_bot = _find_g(p0, "top"), _find_g(p0, "bot")
    l_ang = _get(p0, ["l_star_angle_anchor"], min(l_top, l_bot))
    fp = _get_str(p0, ["grid_fingerprint"])
    dt = _get(p0, ["d_time", "time_grid.dt_sim", "dt_sim"])
    tmax = _get(p0, ["t_max", "time_grid.t_max_sim", "t_max_sim"])
    n_t = int(round(tmax / dt)) if np.isfinite(dt) and dt > 0 else 0

    # verificacion del invariante de la familia: mu_s igualado -> l_s comun
    rel = abs(ls_top - ls_bot) / ls_top
    if rel > 1e-4:
        print(f"[{order}/{pol}] AVISO: l_s NO comun (desviacion relativa {rel:.2e}). "
              f"La regla corta de MM3 supone l_s comun.")
    ls = 0.5 * (ls_top + ls_bot)

    # orden re-derivado de los radios, contra el campo archivado
    order_arch = _get_str(p0, ["order"])
    order_der = "inverse" if r_top > r_bot else "normal"
    if order_arch and order_arch != order_der:
        print(f"[{order}/{pol}] AVISO: order archivado '{order_arch}' != derivado "
              f"'{order_der}' (se usa el derivado)")

    layers = []
    for z in sorted(by_z):
        pz = sw[sorted(by_z[z])[0]].params_flat
        reach = bool(_get(pz, ["reachable_paired"], True))
        layers.append(Layer(z=z, m_star=z / l_top, m_s=z / ls,
                            reachable=reach, keys=sorted(by_z[z])))

    entry_spec = "large" if r_top > r_bot else "small"
    back_spec = "small" if entry_spec == "large" else "large"

    return Campaign(tag=f"{order}/{pol}", order=order_der, pol=pol, sweep=sw,
                    r_entry=r_top, r_back=r_bot, l_entry=l_top, l_back=l_bot,
                    ls_common=ls, g_entry=g_top, g_back=g_bot,
                    l_anchor=l_ang, fingerprint=fp, layers=layers,
                    ctrl={"entry": sorted(ctrl[entry_spec]),
                          "back": sorted(ctrl[back_spec])},
                    dt_sim=dt, t_max_sim=tmax, n_time=n_t)


CAMP = {}
for (order, pol), folder in FOLDERS.items():
    try:
        CAMP[(order, pol)] = load_campaign(order, pol, folder)
        c = CAMP[(order, pol)]
        print(f"[{order:7s}/{pol:8s}] OK  {len(c.layers)} espesores, "
              f"controles: entrada {len(c.ctrl['entry'])} rep, fondo {len(c.ctrl['back'])} rep")
    except Exception as err:
        print(f"[{order:7s}/{pol:8s}] no disponible: {err}")

assert CAMP, "ninguna campania cargada: revisa BASE_PATH y FOLDERS"
LREF = list(CAMP.values())[0].l_anchor          # ancla angular comun
```

```python
# ===========================================================================
# 2c -- tabla del sistema y chequeos de comparabilidad
# ===========================================================================
fps = {c.fingerprint for c in CAMP.values() if c.fingerprint}
print(f"grid_fingerprint distintos: {len(fps)}  -> "
      f"{'restas bin a bin LEGITIMAS' if len(fps) <= 1 else 'AVISO: grillas distintas, interpola antes de restar'}")
for fp in fps:
    print(f"   {fp}")

rows = []
for (order, pol), c in CAMP.items():
    rows.append(dict(order=order, pol=pol,
                     r_entry_nm=1e3 * c.r_entry, r_back_nm=1e3 * c.r_back,
                     l_s_um=c.ls_common,
                     l_star_entry=c.l_entry, l_star_back=c.l_back,
                     g_entry=c.g_entry, g_back=c.g_back,
                     contrast=c.l_back / c.l_entry, R_inf=c.R_inf,
                     dt_sim=c.dt_sim, t_max_sim=c.t_max_sim, n_time=c.n_time))
SYS = pd.DataFrame(rows)
display(SYS.round(4))

c0 = list(CAMP.values())[0]
print(f"\nl_s comun            = {c0.ls_common:.3f} um")
print(f"l* pequena / grande  = {min(c0.l_entry, c0.l_back):.3f} / {max(c0.l_entry, c0.l_back):.3f} um"
      f"   -> contraste {max(c0.l_entry, c0.l_back)/min(c0.l_entry, c0.l_back):.3f}")
print(f"ancla angular LREF   = {LREF:.3f} um   (q = k*LREF*theta)")
print("\nespesores del barrido:")
print("   z [um] | d/l_s | d/l*_normal | d/l*_inverse | alcance temporal")
n_common = min(len(c.layers) for c in CAMP.values())
if n_common != len(c0.layers):
    print(f"AVISO: numero de espesores distinto entre campanias "
          f"({[len(c.layers) for c in CAMP.values()]}); se listan los {n_common} comunes")
for i, L in enumerate(c0.layers[:n_common]):
    ms = {}
    for (o, p), c in CAMP.items():
        ms[o] = c.layers[i].m_star
    print(f"  {L.z:8.2f} | {L.m_s:5.2f} | {ms.get('normal', np.nan):11.2f} | "
          f"{ms.get('inverse', np.nan):12.2f} | {'si' if L.reachable else 'NO':>8}")
print("\n  d/l_s es el MISMO numero en las dos campanias (mu_s igualado);")
print("  d/l*_ent difiere por el contraste. Esa asimetria es el experimento.")
```

```python
# ===========================================================================
# 2d -- resolucion de canales POR FISICA (no por etiqueta)
#   circular : el canal que preserva helicidad es el de apice ~ 2 (Prop. 1.1)
#   lineal   : el co-polarizado es el de mayor apice
# ===========================================================================
def resolve_channel(c):
    if c.pol in CHANNEL_OVERRIDE:
        return CHANNEL_OVERRIDE[c.pol]
    key = c.ctrl["entry"][0] if c.ctrl["entry"] else c.layers[-1].keys[0]
    p1, _ = raw_profiles(c.sweep, key, c.pol, None)
    names = _chan_names(p1)
    apex = {}
    for nm in names:
        try:
            q, E = profile_one(c.sweep, key, c.pol, nm, c.l_anchor)
            apex[nm] = apex_value(q, E)
        except Exception:
            continue
    if not apex:
        raise RuntimeError(f"[{c.tag}] no se pudo leer ningun canal de {names}")
    if c.pol == "circular":
        pick = min(apex, key=lambda nm: abs(apex[nm] - 2.0))
    else:
        pick = max(apex, key=lambda nm: apex[nm])
    txt = "  ".join(f"{nm}: eta(0)={v:.4f}" for nm, v in apex.items())
    print(f"[{c.tag}] canales {names}  ->  {txt}   ELEGIDO: '{pick}'")
    return pick


for c in CAMP.values():
    c.channel = resolve_channel(c)

print("\nCanales en uso:", {t: c.channel for t, c in
                            zip([f"{o}/{p}" for (o, p) in CAMP], CAMP.values())})
print("Si alguno no corresponde a lo esperado, fija CHANNEL_OVERRIDE y re-ejecuta.")
```

```python
# ===========================================================================
# 2e -- perfiles estacionarios (una sola vez; todo lo demas los reutiliza)
# ===========================================================================
PROF = {}     # (order, pol, z)             -> (q, eta)
CTRL = {}     # (order, pol, 'entry'|'back')-> (q, eta)

for (o, p), c in CAMP.items():
    for L in c.layers:
        PROF[(o, p, L.z)] = profile_sum(c.sweep, L.keys, c.pol, c.channel, c.l_anchor)
    for which in ("entry", "back"):
        if c.ctrl[which]:
            CTRL[(o, p, which)] = profile_sum(c.sweep, c.ctrl[which], c.pol,
                                              c.channel, c.l_anchor)
print(f"perfiles estacionarios: {len(PROF)} bicapas, {len(CTRL)} controles")

POL_GATE = "circular" if any(p == "circular" for (_, p) in CAMP) else "linear"
Q_GRID = list(CTRL.values())[0][0] if CTRL else list(PROF.values())[0][0]
print(f"grilla en q: {len(Q_GRID)} bins, dq(apice) = {Q_GRID[1]-Q_GRID[0]:.4f}, "
      f"q_max = {Q_GRID[-1]:.1f}")
```

```python
# ===========================================================================
# 2f -- calibracion del estimador de apice: sesgo contra ruido
#   El unico parametro libre del pipeline es n_apex. Se fija ANTES de mirar
#   fisica, con dos numeros que tiran en direcciones opuestas:
#     sesgo : dispersion residual de una familia SINTETICA de una sola escala
#             sobre la grilla REAL. Deberia ser 0; lo que quede es artefacto
#             del estimador y pone el piso del test de auto-similitud (MM2).
#     ruido : dispersion de eta(0) entre replicas de un control homogeneo.
#   Se elige el menor n cuyo sesgo cae por debajo del ruido.
# ===========================================================================
def synthetic_single_scale(q_grid, l_lo, l_hi, n=12, amp=(0.6, 1.0)):
    """Familia AWM de UNA sola escala sobre el rango real de l*, pasada por el
    MISMO pipeline y la MISMA grilla que los datos."""
    fam = []
    for i, ls in enumerate(np.geomspace(l_lo, l_hi, n)):
        A = amp[0] + (amp[1] - amp[0]) * i / max(n - 1, 1)
        fam.append((np.asarray(q_grid, float),
                    1.0 + A * cone_shape(np.asarray(q_grid, float) * ls / LREF)))
    return fam


L_LO = min(min(c.l_entry, c.l_back) for c in CAMP.values())
L_HI = max(max(c.l_entry, c.l_back) for c in CAMP.values())
SYN = synthetic_single_scale(Q_GRID, L_LO, L_HI)

_ctrl_key = next(iter(CTRL))
_c_ref = CAMP[(_ctrl_key[0], _ctrl_key[1])]
_ctrl_keys = _c_ref.ctrl[_ctrl_key[2]]

print(f"rango de l* cubierto por el piso sintetico: {L_LO:.2f}-{L_HI:.2f} um "
      f"({L_HI/L_LO:.2f}x)")
print("\n   n_apex | sesgo (piso sintetico) | ruido (replicas de eta_0) | S sintetico")
_scan = {}
for n_try in (3, 4, 5, 6, 8, 12):
    _old = CONV["n_apex"]
    CONV["n_apex"] = n_try
    bias = collapse_spread([normalized_shape(qq, EE) for qq, EE in SYN])
    Ss = [shape_invariant(qq, EE) for qq, EE in SYN]
    s_var = 100.0 * (np.nanmax(Ss) / np.nanmin(Ss) - 1.0)
    e0s = [apex_value(*profile_one(_c_ref.sweep, kk, _c_ref.pol,
                                   _c_ref.channel, _c_ref.l_anchor))
           for kk in _ctrl_keys]
    noise = float(np.std(e0s, ddof=1))
    CONV["n_apex"] = _old
    _scan[n_try] = (bias, noise, s_var)
    print(f"   {n_try:6d} | {bias:22.2e} | {noise:25.2e} | {s_var:10.2f} %")

_ok = [n for n, (b, nz, _) in _scan.items() if b < nz]
N_APEX_CAL = min(_ok) if _ok else min(_scan, key=lambda n: _scan[n][0])
print(f"\n   -> n_apex = {N_APEX_CAL}  "
      f"({'sesgo por debajo del ruido de replica' if _ok else 'AVISO: sesgo domina en todo el rango'})")
CONV["n_apex"] = N_APEX_CAL
SYN_FLOOR = _scan[N_APEX_CAL][0]
print(f"   piso del procedimiento para MM2: {SYN_FLOOR:.2e}")
print("   (si el sesgo domina, sube N_THETA_1 en la campania o usa apex_mode='median'\n"
      "    y reporta el piso como limite del test, no como resultado)")
```

## MM0 — Compuertas: nada de lo que sigue es interpretable si esto falla

Dos compuertas independientes, ambas sin parámetros:

1. **Reciprocidad (canal circular).** $\eta_+(0)=2$ está protegido por reciprocidad para dispersores
   con simetría de espejo, *cualquiera sea el medio* (Prop. 1.1). Una bicapa no puede romperlo: si
   $\eta_+(0)$ se mueve con el espesor, el problema es de la maquinaria de capas, no física. Es la
   misma función que cumplió $\eta_+(0)=1.9896$ constante en las mezclas de §5.2.
2. **Límites del barrido.** El espesor mínimo ($d/\ell_s\sim10^{-5}$) debe reproducir el **control
   del fondo**, y el máximo debe acercarse al **control de entrada**. Como las dos campañas
   intercambian los papeles, cada límite homogéneo tiene dos realizaciones independientes: si
   coinciden, el pipeline es consistente en los dos órdenes a la vez.

## MM2 — ¿Admite la bicapa un $\ell^*$ efectivo? Test libre de modelo

Se normaliza cada perfil en amplitud por $\eta(0)-1$ y en ángulo por su propia semianchura:

$$\tilde f(u)=\frac{\eta-1}{\eta(0)-1},\qquad u=\frac{q}{q_{1/2}}.$$

Si un medio admite una única longitud de transporte, su perfil es $1+A\,f(\theta/\Theta)$ para
*alguna* $f$, y $\tilde f$ no depende ni de $A$ ni de $\Theta$: **toda la familia colapsa
exactamente, sea cual sea la forma de línea** (Prop. 4.1). El test no supone AWM ni usa el prefactor
$C$ dependiente de canal.

La lógica exige tres patrones, y aquí los tres están disponibles dentro de la propia campaña:

- **Piso del procedimiento:** familia sintética AWM de una sola escala recorriendo el mismo rango de
  $\ell^*$. Fija la dispersión residual mínima que el procedimiento puede dar.
- **Control positivo medido:** los dos controles homogéneos, que difieren en $\ell^*$ por el factor
  de contraste y **deben** colapsar. Es un control positivo *medido*, con el mismo ruido y la misma
  reducción que los datos.
- **Las bicapas:** si no colapsan, y los dos controles anteriores sí, el no-colapso es un resultado y
  no un defecto del análisis.

El panel (b) da la misma información en forma escalar: el invariante $S=q_{0.25}/q_{0.75}$ es
constante para cualquier medio de una escala. Léase como *dos reglas del mismo cono que en un medio
efectivo tienen que coincidir*.

```python
# ===========================================================================
# MM2 -- auto-similitud: piso sintetico, control medido, bicapas
# ===========================================================================
# --- referencia 1: piso del procedimiento (familia sintetica de una escala) -
SHAPES = {"synthetic": [normalized_shape(qq, EE) for qq, EE in SYN]}
SPREAD = {"synthetic": collapse_spread(SHAPES["synthetic"])}

# --- referencia 2: control positivo MEDIDO (los controles homogeneos) -------
ctrl_shapes = [normalized_shape(*CTRL[kk]) for kk in CTRL]
if len(ctrl_shapes) >= 2:
    SHAPES["controls"] = ctrl_shapes
    SPREAD["controls"] = collapse_spread(ctrl_shapes)

# --- referencia 3: superposicion pura de los dos controles medidos ----------
#   Es la dispersion que tendria una familia que fuese EXACTAMENTE la suma
#   ponderada de los dos medios homogeneos (hipotesis de MM4, sin caminos
#   hibridos). Si las bicapas se quedan en este nivel, el no-colapso queda
#   explicado por la segregacion; si lo superan, hay caminos que visitan las
#   dos capas y no pertenecen a ninguno de los dos terminos.
for (o, p) in [(oo, POL_GATE) for oo in orders if (oo, POL_GATE) in CAMP]:
    if (o, p, "entry") in CTRL and (o, p, "back") in CTRL:
        qref, Ee = CTRL[(o, p, "entry")]
        _,    Eb = CTRL[(o, p, "back")]
        sup = [normalized_shape(qref, 1.0 + w * (Ee - 1.0) + (1.0 - w) * (Eb - 1.0))
               for w in np.linspace(0.0, 1.0, 11)]
        SHAPES["superposition"] = sup
        SPREAD["superposition"] = collapse_spread(sup)
        break

for (o, p), c in CAMP.items():
    tag = f"{o}/{p}"
    SHAPES[tag] = [normalized_shape(*PROF[(o, p, L.z)]) for L in c.layers]
    SPREAD[tag] = collapse_spread(SHAPES[tag])

REFS = ("synthetic", "controls", "superposition")
base = SPREAD.get("controls", SPREAD["synthetic"])
print("[MM2] dispersion del colapso (menor = mas compatible con una sola escala)")
for tagg, v in SPREAD.items():
    extra = "" if tagg in REFS else f"   = {v/base:.1f}x el control medido"
    print(f"      {tagg:20s} {v:.3e}{extra}")
if "superposition" in SPREAD:
    print(f"      -> lectura: piso {SPREAD['synthetic']:.1e} < controles {base:.1e} "
          f"< superposicion {SPREAD['superposition']:.1e} <= bicapas")
    print("         si las bicapas superan la superposicion, hay caminos hibridos")

# --- figura ---------------------------------------------------------------
apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.26})

for s in SHAPES["synthetic"]:
    ax1.plot(U_GRID, s, color="0.75", lw=0.7, zorder=1)
ax1.plot([], [], color="0.75", lw=0.7, label="synthetic single-scale (floor)")
for s in SHAPES.get("controls", []):
    ax1.plot(U_GRID, s, color="k", lw=0.9, ls=":", zorder=3)
if "controls" in SHAPES:
    ax1.plot([], [], color="k", lw=0.9, ls=":", label="homogeneous controls")
for sh in SHAPES.get("superposition", []):
    ax1.plot(U_GRID, sh, color=COL[3], lw=0.6, ls="-.", alpha=0.6, zorder=2)
if "superposition" in SHAPES:
    ax1.plot([], [], color=COL[3], lw=0.9, ls="-.", label="pure superposition")
for (o, p), c in CAMP.items():
    if p != POL_GATE:
        continue
    st = STYLE[o]
    for sh in SHAPES[f"{o}/{p}"]:
        ax1.plot(U_GRID, sh, color=st["color"], lw=0.6, alpha=0.7, zorder=4)
    ax1.plot([], [], color=st["color"], lw=0.9, label=f"bilayer, {o}")
ax1.set_xlabel(r"$u = q/q_{1/2}$")
ax1.set_ylabel(r"$\tilde f(u)$")
ax1.set_title("(a) self-similarity collapse", loc="left")
ax1.set_xlim(0.15, 3.0); ax1.grid(alpha=0.15)
ax1.legend(fontsize=6.5, frameon=False)

# --- (b) invariante de forma ---------------------------------------------
SCAL = {}
for (o, p), c in CAMP.items():
    mu, se = zip(*[scalars_stats(c.sweep, L.keys, c.pol, c.channel, c.l_anchor)
                   for L in c.layers])
    MU, SE = np.asarray(mu), np.asarray(se)
    SCAL[(o, p)] = dict(m_s=c.m_s_arr, m_star=c.m_star_arr, z=c.z_arr,
                        eta0=MU[:, 0], s_eta0=SE[:, 0],
                        rho=MU[:, 1], s_rho=SE[:, 1],
                        qh=MU[:, 2], s_qh=SE[:, 2],
                        S=MU[:, 3], s_S=SE[:, 3])

S_CTRL = []
for kk in CTRL:
    S_CTRL.append(shape_invariant(*CTRL[kk]))
S_CTRL = np.asarray([s for s in S_CTRL if np.isfinite(s)], float)
if len(S_CTRL):
    ax2.axhspan(S_CTRL.min(), S_CTRL.max(), color="0.85", zorder=0,
                label="homogeneous controls")
    print(f"[MM2] S de los controles homogeneos: {S_CTRL.mean():.3f} "
          f"(rango {S_CTRL.min():.3f}-{S_CTRL.max():.3f}, "
          f"{100*(S_CTRL.max()-S_CTRL.min())/S_CTRL.mean():.1f} % pese a variar l* en "
          f"{L_HI/L_LO:.2f}x)")

for (o, p), D in SCAL.items():
    if p != POL_GATE:
        continue
    st = STYLE[o]
    ax2.errorbar(D["m_star"] + 1e-3, D["S"], yerr=D["s_S"], fmt=st["marker"], ms=4,
                 lw=0.9, ls=st["ls"], color=st["color"], capsize=2, label=st["label"])
    if len(S_CTRL):
        dev = 100 * (D["S"] / S_CTRL.mean() - 1)
        i = int(np.nanargmax(np.abs(dev)))
        print(f"[MM2 {o:7s}] desviacion max de S = {dev[i]:+.1f} % en "
              f"d/l*_ent = {D['m_star'][i]:.2f}")

ax2.set_xscale("log")
ax2.set_xlabel(r"$d/\ell^*_{\rm entry}$")
ax2.set_ylabel(r"$S = q_{0.25}/q_{0.75}$")
ax2.set_title("(b) shape invariant", loc="left")
ax2.grid(alpha=0.15, which="both")
ax2.legend(fontsize=6.5, frameon=False)
fig.savefig(FIGDIR / "MM2_autosimilitud.pdf", bbox_inches="tight")
plt.show()
```

## MM3 — Las dos reglas *(figura central del capítulo)*

Ésta es la medición que sólo esta familia permite. A **igual $z$ absoluto**, $d/\ell_s$ coincide entre
órdenes y $d/\ell^*_{\rm ent}$ no. Se normaliza el progreso de cada escalar entre sus propios
límites,

$$P(X)=\frac{X(d)-X(0)}{X(\infty)-X(0)},$$

lo que elimina los valores extremos —distintos entre órdenes, porque las especies de entrada y de
fondo se intercambian— y deja sólo **la escala de profundidad**, que es lo que se quiere medir.

**Panel (a) — regla corta.** Con $S/M=\rho/(1-\rho)\equiv X$ y $M$ prácticamente independiente de $d$
(el fondo difuso a $\theta\approx0$ de un semiespacio sin absorción no depende de $\ell^*$),

$$P(X)=1-e^{-2d/\ell_s}\,,$$

**sin ningún parámetro ajustable** y con $\ell_s$ conocido. Los dos órdenes deben caer sobre la misma
curva aunque el cambio de $\rho$ tenga signos opuestos. Canal lineal: en el circular la dispersión
simple está ruteada fuera y $X\approx0$.

**Panel (b) — regla larga.** El progreso de $q_{1/2}$ contra $d/\ell^*_{\rm ent}$: las dos campañas
deben colapsar. **Panel (c)** el mismo progreso contra $d/\ell_s$ (equivalente a $z$ absoluto): deben
**separarse** por el factor de contraste. La pareja (b)-(c) es el enunciado completo: el ancho mide
$\ell^*$, no $\ell_s$; el ápice mide $\ell_s$, no $\ell^*$.

```python
# ===========================================================================
# MM3a -- regla corta: X = rho/(1-rho) contra d/l_s, prediccion sin parametros
# ===========================================================================
def progress(x):
    x = np.asarray(x, float)
    return (x - x[0]) / (x[-1] - x[0])


RULE_SHORT = {}
for (o, p), D in SCAL.items():
    if p != "linear":
        continue
    rho, s_rho = D["rho"], D["s_rho"]
    X = rho / (1.0 - rho)                       # = S/M, exacto dado M
    sX = s_rho / (1.0 - rho) ** 2               # propagacion lineal
    P = progress(X)
    # error del progreso (extremos correlacionados con el punto: cota superior)
    den = abs(X[-1] - X[0])
    sP = np.sqrt(sX ** 2 + (P ** 2) * sX[-1] ** 2 + ((1 - P) ** 2) * sX[0] ** 2) / den
    RULE_SHORT[o] = dict(m_s=D["m_s"], X=X, sX=sX, P=P, sP=sP,
                         rho=rho, s_rho=s_rho)
    print(f"[MM3a {o:7s}] rho: {rho[0]:.4f} (d->0)  ->  {rho[-1]:.4f} (d->inf)"
          f"   salto {rho[-1]-rho[0]:+.4f}")

if RULE_SHORT:
    ls = list(CAMP.values())[0].ls_common
    mg = np.geomspace(1e-3, max(np.max(v["m_s"]) for v in RULE_SHORT.values()), 300)
    Ppred = 1.0 - np.exp(-2.0 * mg)
    for o, v in RULE_SHORT.items():
        good = v["m_s"] > 1e-3
        resid = v["P"][good] - (1.0 - np.exp(-2.0 * v["m_s"][good]))
        print(f"[MM3a {o:7s}] residuo vs 1-exp(-2d/l_s):  RMS = {np.sqrt(np.mean(resid**2)):.3f}"
              f"   max = {np.max(np.abs(resid)):.3f}   ({np.count_nonzero(good)} espesores)")
    # ajuste libre de la escala, como contraste al valor predicho l_s/2
    for o, v in RULE_SHORT.items():
        good = (v["m_s"] > 1e-3) & (v["P"] < 0.995)
        try:
            f = lambda mm, lam: 1.0 - np.exp(-mm / lam)
            popt, pcov = curve_fit(f, v["m_s"][good], v["P"][good],
                                   p0=[0.5], sigma=np.maximum(v["sP"][good], 1e-3),
                                   absolute_sigma=True, maxfev=20000)
            print(f"[MM3a {o:7s}] escala ajustada = {popt[0]:.3f} +- {np.sqrt(pcov[0,0]):.3f} l_s"
                  f"   (predicha 0.500)")
            RULE_SHORT[o]["lam_fit"] = (float(popt[0]), float(np.sqrt(pcov[0, 0])))
        except Exception as err:
            print(f"[MM3a {o:7s}] ajuste fallo: {err}")
```

```python
# ===========================================================================
# MM3 -- figura de las dos reglas
# ===========================================================================
apply(width_frac=1.0)
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(TEXTWIDTH_IN, 0.36 * TEXTWIDTH_IN),
                                    gridspec_kw={"wspace": 0.32})

# --- (a) regla corta ------------------------------------------------------
if RULE_SHORT:
    ax1.plot(mg, Ppred, "k-", lw=1.0, zorder=1,
             label=r"$1-e^{-2d/\ell_s}$  (no free parameter)")
    for o, v in RULE_SHORT.items():
        st = STYLE[o]
        ax1.errorbar(v["m_s"], v["P"], yerr=v["sP"], fmt=st["marker"], ms=4.5,
                     lw=0, elinewidth=0.9, color=st["color"], capsize=2,
                     label=st["label"], zorder=3)
ax1.set_xscale("log"); ax1.set_xlim(1e-2, None); ax1.set_ylim(-0.08, 1.12)
ax1.set_xlabel(r"$d/\ell_s$")
ax1.set_ylabel(r"progress of $X=\rho/(1-\rho)$")
ax1.set_title(r"(a) short ruler: apex, scale $\ell_s/2$", loc="left", fontsize="small")
ax1.grid(alpha=0.15, which="both")
ax1.legend(fontsize=6, frameon=False, loc="lower right")

# --- (b) regla larga, unidades de transporte ------------------------------
for (o, p), D in SCAL.items():
    if p != POL_GATE:
        continue
    st = STYLE[o]
    P = progress(D["qh"])
    ax2.plot(D["m_star"], P, st["ls"], marker=st["marker"], ms=4, lw=0.9,
             color=st["color"], label=st["label"])
ax2.set_xscale("log")
ax2.set_xlabel(r"$d/\ell^*_{\rm entry}$")
ax2.set_ylabel(r"progress of $q_{1/2}$")
ax2.set_title(r"(b) long ruler: collapses in $\ell^*$", loc="left", fontsize="small")
ax2.grid(alpha=0.15, which="both")
ax2.legend(fontsize=6, frameon=False)

# --- (c) el mismo progreso contra la regla corta: debe fallar --------------
for (o, p), D in SCAL.items():
    if p != POL_GATE:
        continue
    st = STYLE[o]
    ax3.plot(D["m_s"], progress(D["qh"]), st["ls"], marker=st["marker"], ms=4,
             lw=0.9, color=st["color"], label=st["label"])
ax3.set_xscale("log")
ax3.set_xlabel(r"$d/\ell_s$  (equivalently absolute $z$)")
ax3.set_title(r"(c) same, wrong ruler", loc="left", fontsize="small")
ax3.grid(alpha=0.15, which="both")

fig.savefig(FIGDIR / "MM3_dos_reglas.pdf", bbox_inches="tight")
plt.show()

# --- numero citable: discrepancia entre ordenes en cada normalizacion ------
def overlap_rms(xa, Pa, xb, Pb, n=40):
    lo, hi = max(np.min(xa), np.min(xb)), min(np.max(xa), np.max(xb))
    if not (hi > lo):
        return np.nan
    g = np.geomspace(max(lo, 1e-6), hi, n)
    d = np.interp(g, xa, Pa) - np.interp(g, xb, Pb)
    return float(np.sqrt(np.mean(d ** 2)))

if len(orders) == 2 and all((o, POL_GATE) in SCAL for o in orders):
    A, B = SCAL[(orders[0], POL_GATE)], SCAL[(orders[1], POL_GATE)]
    PA, PB = progress(A["qh"]), progress(B["qh"])
    r_star = overlap_rms(A["m_star"], PA, B["m_star"], PB)
    r_s = overlap_rms(A["m_s"], PA, B["m_s"], PB)
    print(f"\n[MM3] discrepancia entre ordenes del progreso de q_1/2:")
    print(f"      contra d/l*_ent : RMS = {r_star:.3f}   <- colapsa")
    print(f"      contra d/l_s    : RMS = {r_s:.3f}   <- no colapsa")
    print(f"      razon = {r_s/r_star:.1f}x   (el ancho mide l*, no l_s)")
```

## MM4 — Cono único contra superposición de los dos controles

Dos modelos enfrentados sobre la **misma** ventana de ajuste:

1. **Cono único** (hipótesis de medio efectivo, 2 parámetros $A,\ \ell^*_{\rm eff}$):
   $\eta=1+A\,\mathcal C(k\ell^*_{\rm eff}\theta)$. Si la bicapa admitiera una escala, esto ajustaría
   tan bien como ajusta a los controles homogéneos. El punto de comparación honesto es precisamente
   el RMS que este mismo ajuste deja sobre los controles: es el piso instrumental.
2. **Superposición de los dos controles medidos** (1 parámetro $w$ por espesor, regresión cerrada):
   $\eta_d-1=w\,[\eta_{\rm ent}-1]+(1-w)\,[\eta_{\rm fondo}-1]$. Usa perfiles medidos, no una forma de
   línea supuesta. El residuo que quede mide los **caminos híbridos**, los que visitan las dos capas
   y no pertenecen a ninguno de los dos términos.

$w(m)$ es la fracción de peso que la capa de entrada se lleva. Por la estadística de excursiones
brownianas su saturación es de **cola pesada**, $w=m/(m+\zeta)$ con $\zeta\approx1+\tilde b=1.71$ en
unidades de $\ell^*_{\rm ent}$; se contrasta contra la alternativa exponencial, que debe perder.

```python
# ===========================================================================
# MM4 -- cono unico vs dos conos; w(m)
# ===========================================================================
def fit_single_cone(q, E, q_max=CONV["q_fit"]):
    """eta = 1 + A * C(q * L/LREF). Devuelve (A, L_eff [um], RMS)."""
    m = q <= q_max
    model = lambda qq, A, L: 1.0 + A * cone_shape(qq * L / LREF)
    p0 = [max(np.max(E[m]) - 1.0, 0.1), LREF]
    popt, _ = curve_fit(model, q[m], E[m], p0=p0, maxfev=40000)
    resid = E[m] - model(q[m], *popt)
    return float(popt[0]), float(popt[1]), float(np.sqrt(np.mean(resid ** 2)))


def fit_two_cones(q, E, E_entry, E_back, q_max=CONV["q_fit"]):
    """w cerrado por minimos cuadrados sobre (y - f_back) = w (f_entry - f_back)."""
    m = q <= q_max
    ft, fb, y = E_entry - 1.0, E_back - 1.0, E - 1.0
    dd = (ft - fb)[m]
    w = float(np.sum((y - fb)[m] * dd) / np.sum(dd ** 2))
    resid = (y - fb - w * (ft - fb))[m]
    return w, float(np.sqrt(np.mean(resid ** 2)))


TWO = {}
for (o, p), c in CAMP.items():
    if p != POL_GATE or (o, p, "entry") not in CTRL or (o, p, "back") not in CTRL:
        continue
    q_ref, E_entry = CTRL[(o, p, "entry")]
    _,     E_back  = CTRL[(o, p, "back")]

    # piso instrumental: el mismo ajuste de cono unico sobre los controles
    floor = [fit_single_cone(*CTRL[(o, p, wch)])[2] for wch in ("entry", "back")]
    floor = float(np.mean(floor))

    rows = []
    for L in c.layers:
        q, E = PROF[(o, p, L.z)]
        Eb = E if np.allclose(q, q_ref) else np.interp(q_ref, q, E)
        qq = q if np.allclose(q, q_ref) else q_ref
        A1, Leff, rms1 = fit_single_cone(qq, Eb)
        w, rms2 = fit_two_cones(qq, Eb, E_entry, E_back)
        ws = []
        for kk in L.keys:
            qr, Er = profile_one(c.sweep, kk, c.pol, c.channel, c.l_anchor)
            Er = Er if np.allclose(qr, q_ref) else np.interp(q_ref, qr, Er)
            ws.append(fit_two_cones(q_ref, Er, E_entry, E_back)[0])
        sw = t_ci(len(ws)) * np.std(ws, ddof=1) / np.sqrt(len(ws))
        rows.append(dict(m_star=L.m_star, m_s=L.m_s, z=L.z, L_eff=Leff,
                         rms_single=rms1, w=w, s_w=sw, rms_two=rms2))
    TWO[o] = pd.DataFrame(rows)
    TWO[o].attrs["floor"] = floor
    print(f"[MM4 {o:7s}] piso instrumental (cono unico sobre controles) RMS = {floor:.2e}")
    print(f"           cono unico sobre bicapas: RMS max = {TWO[o]['rms_single'].max():.2e}"
          f"  ({TWO[o]['rms_single'].max()/floor:.1f}x el piso)")
    print(f"           dos conos:                RMS max = {TWO[o]['rms_two'].max():.2e}"
          f"  ({TWO[o]['rms_two'].max()/floor:.1f}x el piso)")
```

```python
# ===========================================================================
# MM4 -- figura y ajuste de w(m)
# ===========================================================================
def w_heavy(m_, zeta):  return m_ / (m_ + zeta)
def w_exp(m_, zeta):    return 1.0 - np.exp(-m_ / zeta)

apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.28})

WFIT = {}
for o, df in TWO.items():
    st = STYLE[o]
    ax1.plot(df["m_star"], df["rms_single"], st["ls"], marker=st["marker"], ms=4,
             lw=0.9, color=st["color"], label=f"{o}: single cone")
    ax1.plot(df["m_star"], df["rms_two"], st["ls"], marker="^", ms=4, lw=0.9,
             color=st["color"], alpha=0.55, mfc="none", label=f"{o}: two cones")
    ax1.axhline(df.attrs["floor"], color=st["color"], lw=0.6, ls=":")

    good = np.isfinite(df["w"]) & (df["m_star"] > 1e-3)
    mm, ww = df["m_star"][good].values, df["w"][good].values
    sww = np.maximum(df["s_w"][good].values, 1e-3)
    res = {}
    for name, fn, p0 in (("heavy", w_heavy, ZETA_TEO), ("exp", w_exp, 1.0)):
        try:
            popt, pcov = curve_fit(fn, mm, ww, p0=[p0], sigma=sww,
                                   absolute_sigma=True, maxfev=20000)
            chi2 = float(np.sum(((ww - fn(mm, *popt)) / sww) ** 2))
            res[name] = (float(popt[0]), float(np.sqrt(pcov[0, 0])), chi2)
        except Exception as err:
            print(f"[MM4 {o}] ajuste {name} fallo: {err}")
    WFIT[o] = res
    if "heavy" in res:
        z_, sz_, c2 = res["heavy"]
        print(f"[MM4 {o:7s}] w = m/(m+zeta):  zeta = {z_:.3f} +- {sz_:.3f}"
              f"   (teoria 1+b = {ZETA_TEO:.3f})   chi2/ndf = {c2:.1f}/{len(mm)-1}")
    if "exp" in res:
        print(f"[MM4 {o:7s}] alternativa exponencial: chi2/ndf = "
              f"{res['exp'][2]:.1f}/{len(mm)-1}"
              f"   -> {'la cola pesada gana' if res.get('heavy', (0,0,1e9))[2] < res['exp'][2] else 'REVISAR'}")

    ax2.errorbar(df["m_star"], df["w"], yerr=df["s_w"], fmt=st["marker"], ms=4,
                 lw=0, elinewidth=0.9, color=st["color"], capsize=2, label=st["label"])
    if "heavy" in res:
        mgrid = np.geomspace(max(df["m_star"].min(), 1e-3), df["m_star"].max(), 200)
        ax2.plot(mgrid, w_heavy(mgrid, res["heavy"][0]), st["ls"], lw=0.9,
                 color=st["color"], alpha=0.8)

mgrid = np.geomspace(1e-2, 20, 200)
ax2.plot(mgrid, w_heavy(mgrid, ZETA_TEO), "k-", lw=0.8, alpha=0.7,
         label=rf"$m/(m+\zeta)$, $\zeta=1+\tilde b={ZETA_TEO:.2f}$")

ax1.set_xscale("log"); ax1.set_yscale("log")
ax1.set_xlabel(r"$d/\ell^*_{\rm entry}$"); ax1.set_ylabel("fit RMS")
ax1.set_title("(a) effective medium vs superposition", loc="left", fontsize="small")
ax1.grid(alpha=0.15, which="both"); ax1.legend(fontsize=6, frameon=False)

ax2.set_xscale("log"); ax2.set_ylim(-0.05, 1.15)
ax2.set_xlabel(r"$d/\ell^*_{\rm entry}$"); ax2.set_ylabel(r"entry weight $w$")
ax2.set_title("(b) weight of the entry layer", loc="left", fontsize="small")
ax2.grid(alpha=0.15, which="both"); ax2.legend(fontsize=6, frameon=False)

fig.savefig(FIGDIR / "MM4_dos_conos.pdf", bbox_inches="tight")
plt.show()
```

## MM5 — El orden es un observable

Las dos campañas comparten especies, fracciones de volumen, profundidades absolutas de interfaz y
—por el `grid_fingerprint`— la grilla bin a bin. Lo único que cambia es cuál medio recibe la luz. En
esta familia, además, **$\mu_s$ es igual en todas partes**: las dos muestras contienen exactamente la
misma cantidad de dispersión por unidad de longitud, sólo reordenada en profundidad.

$$\Delta_{\rm orden}(q;z)=\eta_{\rm normal}(q;z)-\eta_{\rm inverse}(q;z)$$

es entonces una afirmación más fuerte que la ausencia de medio efectivo: dice que el observable
depende del **orden de apilamiento**, no sólo de qué hay y en qué proporción. Escala de referencia:
la brecha de Jensen de §5.2 (mezclas binarias) vale $\sim4\times10^{-3}$; ahí el desorden
composicional era conmutativo y el residuo era de segundo orden. Aquí no lo es.

## MM6 — Cono con compuerta temporal: la interfaz enterrada

La resolución temporal selecciona clases de camino por su longitud $L=vt$ en lugar de integrarlas.
Dentro de un medio homogéneo el cono se angosta como $t^{-1/2}$ y, a $t$ fijo, su ancho escala como
$(\ell^*)^{-1/2}$ (ec. 45) — nótese el cambio de exponente respecto del estacionario, donde la
integración sobre $t$ convierte $(\ell^*)^{-1/2}$ en $(\ell^*)^{-1}$.

Con contraste $\ell^*$ de sólo 1.8, el rango dinámico total del efecto es
$\sqrt{1.8}=1.34$: el cociente crudo $R(t)$ usado en §5.3 es demasiado poco sensible. Se usa en su
lugar la **coordenada de interpolación**, que normaliza por *todo* el contraste disponible y por eso
recupera la sensibilidad:

$$\lambda(t)=\frac{\ln q_{1/2}^{\rm bicapa}(t)-\ln q_{1/2}^{\rm entrada}(t)}
                  {\ln q_{1/2}^{\rm fondo}(t)-\ln q_{1/2}^{\rm entrada}(t)}\;\in[0,1],$$

construida con los **dos controles medidos** en la misma grilla temporal: $\lambda\to0$ mientras la
nube difusiva vive en la capa de entrada, $\lambda\to1$ cuando ha cruzado. Además cancela a primer
orden los efectos de piso pre-difusivos de los primeros bins, que afectan por igual a los tres
perfiles. El tiempo de quiebre es $\lambda(t^*)=1/2$, y la predicción sin parámetros es

$$t^*=\tfrac32\,\Big(\frac{d}{\ell^*_{\rm ent}}\Big)^{2}\,\tau^*_{\rm ent}.$$

**Advertencia metodológica.** Los primeros bins temporales son balísticos/serpenteantes: no obedecen
los escalamientos difusivos y se excluyen del ajuste de $t^*$. Sólo entran los espesores cuya
interfaz es alcanzable dentro de la ventana temporal (`reachable`).

```python
# ===========================================================================
# MM6a -- anchos con compuerta temporal
# ===========================================================================
TIME_STRIDE = 1
POL_TIME = "circular" if any(p == "circular" for (_, p) in CAMP) else "linear"


def noise_tail(q, E, q_min=CONV["q_view"]):
    m = q > q_min
    return float(np.std(E[m])) if np.count_nonzero(m) > 5 else np.inf


def gated_width(c, keys):
    """(t/tau*_ent, q_1/2) por bin temporal, con compuerta de S/N."""
    ts, ws = [], []
    for it in range(1, c.n_time + 1, TIME_STRIDE):
        try:
            q, E = profile_sum(c.sweep, keys, c.pol, c.channel, c.l_anchor,
                               time_index=int(it))
        except Exception:
            continue
        e0 = apex_value(q, E)
        if not np.isfinite(e0) or (e0 - 1.0) < CONV["snr_min_time"] * noise_tail(q, E):
            continue
        h = level_q(q, E, 0.5)
        if np.isfinite(h) and h > 0:
            ts.append((it - 0.5) * c.dt_sim / c.tau_entry)
            ws.append(h)
    return np.asarray(ts), np.asarray(ws)


WID, WCTL = {}, {}
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    for which in ("entry", "back"):
        if c.ctrl[which]:
            WCTL[(o, which)] = gated_width(c, c.ctrl[which])
    for L in c.layers:
        WID[(o, L.z)] = gated_width(c, L.keys)
    print(f"[MM6 {o:7s}] bins validos -- controles: "
          f"entrada {len(WCTL.get((o,'entry'), ([],[]))[0])}, "
          f"fondo {len(WCTL.get((o,'back'), ([],[]))[0])};  "
          f"bicapas: {[len(WID[(o, L.z)][0]) for L in c.layers]}")
```

```python
# ===========================================================================
# MM6b -- lambda(t), t* y el cambio de signo del plateau
# ===========================================================================
def lam_curve(o, z):
    """Coordenada de interpolacion entre los dos controles, en la malla comun."""
    te, we = WCTL.get((o, "entry"), (np.array([]), np.array([])))
    tb, wb = WCTL.get((o, "back"), (np.array([]), np.array([])))
    td, wd = WID.get((o, z), (np.array([]), np.array([])))
    if min(len(te), len(tb), len(td)) < 8:
        return np.array([]), np.array([])
    lo = max(te.min(), tb.min(), td.min())
    hi = min(te.max(), tb.max(), td.max())
    if not (hi > lo):
        return np.array([]), np.array([])
    g = np.geomspace(lo, hi, 60)
    le = np.log(np.interp(g, te, we)); lb = np.log(np.interp(g, tb, wb))
    ld = np.log(np.interp(g, td, wd))
    den = lb - le
    with np.errstate(divide="ignore", invalid="ignore"):
        lam = np.where(np.abs(den) > 1e-6, (ld - le) / den, np.nan)
    return g, lam


def first_crossing(x, y, level=0.5, persist=2):
    """Primer cruce ascendente persistente (evita cruces por ruido de un bin)."""
    above = y > level
    for i in range(1, len(y) - persist):
        if (not above[i - 1]) and np.all(above[i:i + persist]):
            if y[i] == y[i - 1]:
                return float(x[i])
            return float(x[i - 1] + (level - y[i - 1]) * (x[i] - x[i - 1]) / (y[i] - y[i - 1]))
    return np.nan


apply(width_frac=1.0)
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(TEXTWIDTH_IN, 0.36 * TEXTWIDTH_IN),
                                    gridspec_kw={"wspace": 0.34})

TSTAR = {}
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    st = STYLE[o]
    # (a) anchos crudos + controles
    for which, lsty in (("entry", "-"), ("back", "--")):
        if (o, which) in WCTL:
            t_, w_ = WCTL[(o, which)]
            ax1.plot(t_, w_, lsty, color=st["color"], lw=1.4, alpha=0.9,
                     label=f"{o}: {which} control")
    cmap = plt.cm.viridis(np.linspace(0, 0.92, len(c.layers)))
    for col, L in zip(cmap, c.layers):
        t_, w_ = WID[(o, L.z)]
        if len(t_):
            ax1.plot(t_, w_, "-", color=col, lw=0.6, alpha=0.75)

    # (b) lambda(t) y t*
    rows = []
    for col, L in zip(cmap, c.layers):
        g, lam = lam_curve(o, L.z)
        if not len(g):
            continue
        ax2.plot(g, lam, "-", color=col, lw=0.7, alpha=0.85)
        if not L.reachable:
            continue
        ts = first_crossing(g, lam, 0.5)
        lam_late = float(np.nanmedian(lam[-max(3, len(lam)//6):]))
        rows.append(dict(m_star=L.m_star, m_s=L.m_s, z=L.z, t_star=ts,
                         lam_late=lam_late))
    TSTAR[o] = pd.DataFrame(rows)

    # (c) t* vs m^2
    df = TSTAR[o].dropna(subset=["t_star"])
    if len(df) >= 2:
        u, t = df["m_star"].values ** 2, df["t_star"].values
        a_fit = float(np.sum(u * t) / np.sum(u ** 2))
        ax3.plot(u, t, st["marker"], ms=5, color=st["color"],
                 label=rf"{o}, $\alpha={a_fit:.2f}$")
        TSTAR[o].attrs["alpha"] = a_fit
        print(f"[MM6 {o:7s}] alpha ajustado = {a_fit:.2f}  (teoria {ALPHA_TEO})"
              f"   con {len(df)} espesores alcanzables")
        print(f"           lambda tardio: {np.round(df['lam_late'].values, 2)}"
              f"   (1.0 = cono del fondo alcanzado)")
    else:
        print(f"[MM6 {o:7s}] insuficientes cruces resolubles de lambda")

ax1.set_xscale("log"); ax1.set_yscale("log")
ax1.set_xlabel(r"$t/\tau^*_{\rm entry}$"); ax1.set_ylabel(r"$q_{1/2}(t)$")
ax1.set_title("(a) gated cone width", loc="left", fontsize="small")
ax1.grid(alpha=0.15, which="both"); ax1.legend(fontsize=6, frameon=False)

ax2.axhline(0, c="k", lw=0.5); ax2.axhline(1, c="k", lw=0.5)
ax2.axhline(0.5, c="k", lw=0.6, ls=":")
ax2.set_xscale("log"); ax2.set_ylim(-0.3, 1.4)
ax2.set_xlabel(r"$t/\tau^*_{\rm entry}$"); ax2.set_ylabel(r"$\lambda(t)$")
ax2.set_title("(b) interpolation coordinate", loc="left", fontsize="small")
ax2.grid(alpha=0.15, which="both")

uu = np.geomspace(1e-3, 1e2, 50)
ax3.plot(uu, ALPHA_TEO * uu, "k-", lw=0.8, alpha=0.7, label=r"$\alpha=3/2$")
ax3.set_xscale("log"); ax3.set_yscale("log")
ax3.set_xlabel(r"$(d/\ell^*_{\rm entry})^2$"); ax3.set_ylabel(r"$t^*/\tau^*_{\rm entry}$")
ax3.set_title("(c) diffusive depth scaling", loc="left", fontsize="small")
ax3.grid(alpha=0.15, which="both"); ax3.legend(fontsize=6, frameon=False)

fig.savefig(FIGDIR / "MM6_tiempo.pdf", bbox_inches="tight")
plt.show()

# --- cambio de signo del plateau, contra la prediccion sin parametros ------
print("\n[MM6] plateau tardio de anchos (cambio de signo predicho):")
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    te, we = WCTL.get((o, "entry"), (np.array([]), np.array([])))
    tb, wb = WCTL.get((o, "back"), (np.array([]), np.array([])))
    if len(te) > 5 and len(tb) > 5:
        lo, hi = max(te.min(), tb.min()), min(te.max(), tb.max())
        g = np.geomspace(lo, hi, 40)[-10:]
        R_meas = float(np.median(np.interp(g, tb, wb) / np.interp(g, te, we)))
        print(f"      {o:7s}: R_inf medido (fondo/entrada, tardio) = {R_meas:.3f}"
              f"   predicho sqrt(l*_ent/l*_fondo) = {c.R_inf:.3f}"
              f"   -> el cono {'se ensancha' if c.R_inf > 1 else 'se estrecha'}")
```

## MM7 — Inversión: espesor y medio enterrado

Tres inversiones, cada una con su rango de validez explícito. Se hacen **a ciegas**: se usa sólo lo
que un experimentador tendría (el medio de entrada caracterizado, $\ell_s$ conocido por ser $\mu_s$
igualado) y se compara al final con el valor verdadero del barrido.

**(1) Espesor por la regla corta.** De $P(X)=1-e^{-2d/\ell_s}$,
$$\hat d=-\tfrac{\ell_s}{2}\ln\!\big(1-P(X)\big),\qquad
\delta\hat d=\frac{\ell_s}{2}\,\frac{\delta P}{1-P}.$$
Útil mientras $P$ no sature, es decir $d\lesssim2\ell_s$; la incertidumbre diverge al saturar, y ése
es el límite honesto del método, no un fallo.

**(2) Espesor por la regla larga.** De $t^*=\tfrac32 m^2\tau^*_{\rm ent}$,
$\hat d=\ell^*_{\rm ent}\sqrt{2t^*/(3\tau^*_{\rm ent})}$. Toma el relevo justo donde la regla corta
satura: el rango útil empieza en $d\sim\ell^*$.

**(3) El medio enterrado.** Del plateau tardío, $\ell^*_{\rm fondo}=\ell^*_{\rm ent}/R_\infty^2$; y
como $\mu_s$ está igualado y $\ell_s$ es conocido,
$$g_{\rm fondo}=1-\frac{\ell_s}{\ell^*_{\rm fondo}}\;\longrightarrow\;a_{\rm fondo}\ \text{por Mie}.$$
Esta última cadena es privativa de la familia $\mu_s$ igualado: sin ese invariante, $\ell^*$ solo no
separa densidad de tamaño.

```python
# ===========================================================================
# MM7 -- inversion ciega de espesor
# ===========================================================================
INV = []
ls_c = list(CAMP.values())[0].ls_common

# --- (1) regla corta ------------------------------------------------------
for o, v in RULE_SHORT.items():
    for i in range(len(v["m_s"])):
        P, sP = v["P"][i], v["sP"][i]
        if not (0.0 < P < 0.98):
            continue
        d_hat = -0.5 * ls_c * np.log(max(1.0 - P, 1e-6))
        s_d = 0.5 * ls_c * sP / max(1.0 - P, 1e-6)
        INV.append(dict(order=o, route="apex ($\\ell_s$)", z_true=v["m_s"][i] * ls_c,
                        d_hat=d_hat, s_d=s_d))

# --- (2) regla larga ------------------------------------------------------
for o, df in TSTAR.items():
    c = CAMP[(o, POL_TIME)]
    for _, r in df.dropna(subset=["t_star"]).iterrows():
        d_hat = c.l_entry * np.sqrt(2.0 * r["t_star"] / (3.0))
        INV.append(dict(order=o, route="gated ($t^*$)", z_true=r["z"],
                        d_hat=d_hat, s_d=np.nan))

DF_INV = pd.DataFrame(INV)

apply(width_frac=1.0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.42 * TEXTWIDTH_IN),
                               gridspec_kw={"wspace": 0.30})

if len(DF_INV):
    zz = np.geomspace(max(DF_INV["z_true"].min(), 1e-3), DF_INV["z_true"].max(), 50)
    ax1.plot(zz, zz, "k-", lw=0.8, alpha=0.7, label="identity")
    for (o, route), sub in DF_INV.groupby(["order", "route"]):
        st = STYLE[o]
        mk = st["marker"] if "apex" in route else "^"
        ax1.errorbar(sub["z_true"], sub["d_hat"],
                     yerr=sub["s_d"] if sub["s_d"].notna().any() else None,
                     fmt=mk, ms=4.5, lw=0, elinewidth=0.8, capsize=2,
                     color=st["color"], mfc="none" if "gated" in route else None,
                     label=f"{o}, {route}")
    ax1.set_xscale("log"); ax1.set_yscale("log")
    ax1.set_xlabel(r"true $d$ [$\mu$m]"); ax1.set_ylabel(r"recovered $\hat d$ [$\mu$m]")
    ax1.set_title("(a) blind thickness inversion", loc="left", fontsize="small")
    ax1.grid(alpha=0.15, which="both"); ax1.legend(fontsize=6, frameon=False)

    DF_INV["rel_err"] = DF_INV["d_hat"] / DF_INV["z_true"] - 1.0
    print("[MM7] error relativo de la inversion de espesor:")
    for (o, route), sub in DF_INV.groupby(["order", "route"]):
        print(f"      {o:7s} {route:16s}: mediana {100*sub['rel_err'].median():+.1f} %"
              f"   |err| max {100*sub['rel_err'].abs().max():.1f} %"
              f"   ({len(sub)} puntos, d in [{sub['z_true'].min():.2f}, {sub['z_true'].max():.1f}] um)")

# --- (3) medio enterrado --------------------------------------------------
BURIED = []
for (o, p), c in CAMP.items():
    if p != POL_TIME:
        continue
    te, we = WCTL.get((o, "entry"), (np.array([]), np.array([])))
    tb, wb = WCTL.get((o, "back"), (np.array([]), np.array([])))
    if len(te) < 6 or len(tb) < 6:
        continue
    lo, hi = max(te.min(), tb.min()), min(te.max(), tb.max())
    g = np.geomspace(lo, hi, 40)[-10:]
    R = float(np.median(np.interp(g, tb, wb) / np.interp(g, te, we)))
    l_back_hat = c.l_entry / R ** 2
    g_back_hat = 1.0 - ls_c / l_back_hat
    BURIED.append(dict(order=o, R_meas=R, R_pred=c.R_inf,
                       l_back_hat=l_back_hat, l_back_true=c.l_back,
                       g_back_hat=g_back_hat, g_back_true=c.g_back))
DF_BUR = pd.DataFrame(BURIED)
if len(DF_BUR):
    display(DF_BUR.round(4))
    idx = np.arange(len(DF_BUR))
    ax2.bar(idx - 0.18, DF_BUR["l_back_hat"], 0.34, label=r"recovered $\ell^*_{\rm back}$",
            color=COL[2])
    ax2.bar(idx + 0.18, DF_BUR["l_back_true"], 0.34, label=r"true $\ell^*_{\rm back}$",
            color="0.75")
    ax2.set_xticks(idx); ax2.set_xticklabels(DF_BUR["order"])
    ax2.set_ylabel(r"$\ell^*$ [$\mu$m]")
    ax2.set_title("(b) buried medium recovery", loc="left", fontsize="small")
    ax2.grid(alpha=0.15, axis="y"); ax2.legend(fontsize=6, frameon=False)
    for _, r in DF_BUR.iterrows():
        print(f"[MM7 {r['order']:7s}] l*_fondo: recuperado {r['l_back_hat']:.2f} um vs "
              f"verdadero {r['l_back_true']:.2f} um  ({100*(r['l_back_hat']/r['l_back_true']-1):+.1f} %)")
        print(f"           g_fondo:  recuperado {r['g_back_hat']:.4f} vs "
              f"verdadero {r['g_back_true']:.4f}")

fig.savefig(FIGDIR / "MM7_inversion.pdf", bbox_inches="tight")
plt.show()
```

```python
# ===========================================================================
# MM7c -- del g recuperado al RADIO enterrado (opcional: requiere luminis_mc)
# ===========================================================================
try:
    from luminis_mc import (MiePhaseFunction, MieMedium,
                            RayleighDebyeEMCPhaseFunction, RGDMedium,
                            derived_quantities)

    N_PARTICLE = 1.59
    _keep = []

    def g_of_radius(a, model="mie", phi=0.10):
        if model == "mie":
            ph = MiePhaseFunction(WAVELENGTH, a, N_PARTICLE, N_MEDIUM, 4000, 0.0, np.pi)
            med = MieMedium(ph, a, N_PARTICLE, N_MEDIUM, WAVELENGTH)
        else:
            ph = RayleighDebyeEMCPhaseFunction(WAVELENGTH, a, N_PARTICLE, N_MEDIUM,
                                               4000, 0.0, np.pi)
            med = RGDMedium(ph, a, N_PARTICLE, N_MEDIUM, WAVELENGTH)
        _keep.append(ph)
        return float(derived_quantities(med, phi)["anisotropy_g"])

    radii = np.geomspace(0.02, 0.20, 40)
    gg = np.array([g_of_radius(a) for a in radii])
    for _, r in DF_BUR.iterrows():
        c = CAMP[(r["order"], POL_TIME)]
        if not np.isfinite(r["g_back_hat"]):
            continue
        j = int(np.argmin(np.abs(gg - r["g_back_hat"])))
        a_hat = float(np.interp(r["g_back_hat"], gg, radii)) if np.all(np.diff(gg) > 0) else radii[j]
        print(f"[MM7c {r['order']:7s}] radio enterrado: recuperado {1e3*a_hat:.1f} nm  "
              f"vs verdadero {1e3*c.r_back:.1f} nm  ({100*(a_hat/c.r_back-1):+.1f} %)")
    print("      cadena: R_inf -> l*_fondo -> g_fondo = 1 - l_s/l*_fondo -> a  "
          "(solo posible con mu_s igualado)")
except Exception as err:
    print(f"[MM7c] inversion a radio no ejecutada ({err}).")
    print("       Con l_s conocido, g_fondo = 1 - l_s/l*_fondo ya esta en DF_BUR;")
    print("       basta invertir g(a) con una tabla de Mie para la especie candidata.")
```

## Notas, límites y cosas que hay que verificar antes de citar

**Lo que esta campaña puede afirmar y la de §5.3 no.** Que el ancho del cono mide $\ell^*$ y el ápice
mide $\ell_s$ es, en general, una afirmación confundida: al variar la especie cambian las dos
longitudes a la vez. Con $\mu_s$ igualado, $\ell_s$ queda fijo y las dos campañas comparten
$d/\ell_s$ a $z$ igual: la separación es limpia y **sin parámetros ajustados**. Es el argumento más
fuerte del capítulo y conviene que MM3 sea la figura que lo lleva.

**Lo que esta campaña afirma peor.** Con contraste $\ell^*\simeq1.8$ (frente a 5.09), el perfil de dos
escalas no se separa a la vista y el rango dinámico temporal es $\sqrt{1.8}=1.34$. Por eso:
- MM1 es descriptiva; el no-colapso hay que sostenerlo con MM2 y MM4, con el piso instrumental
  explícito, no con la figura.
- MM6 usa $\lambda(t)$ y no el $R(t)$ crudo de §5.3. Si $\lambda$ no alcanza $1/2$ de forma
  persistente para un espesor dado, ese punto **no entra** al ajuste de $\alpha$; es preferible
  reportar pocos puntos limpios que forzar cruces por ruido.

**Verificaciones obligatorias antes de citar cualquier número.**
1. Que la celda de resolución de canales eligió lo esperado. En circular, el canal elegido debe dar
   $\eta(0)\approx2$ en los controles homogéneos; si no, hay un intercambio de etiqueta y hay que
   fijar `CHANNEL_OVERRIDE`.
2. Que el `grid_fingerprint` es único entre las cuatro campañas. Si no lo es, MM5 no puede restar bin
   a bin y hay que interpolar (la celda avisa).
3. Que no aparecen avisos de `[stitch WARN]`: el artefacto de costura en $|q|\approx7.5$ documentado
   en §5.1 caería justo dentro de la ventana de cola usada para la línea base.
4. Que la lista de espesores `reachable` coincide con lo que imprimió el script generador. Los
   espesores no alcanzables dentro de la ventana temporal siguen siendo válidos en estacionario, pero
   quedan fuera de MM6 y MM7-ruta-2 por construcción.

**Supuesto explícito de la regla corta.** $P(X)=1-e^{-2d/\ell_s}$ asume que el fondo múltiple $M$ a
$\theta\approx0$ no depende de $d$. Es buena aproximación en un semiespacio sin absorción, donde toda
la luz retorna y la distribución angular del albedo difuso es prácticamente independiente de
$\ell^*$; el residuo impreso en MM3a **es** la medida de esa aproximación y hay que reportarlo, no
esconderlo. Si el residuo crece de forma sistemática con $d$, el enunciado correcto es que la escala
del ápice es $\ell_s/2$ *hasta* la corrección difusiva, cuya magnitud es el número que se cita.

**Coherencia con §5.3.** Los observables ($\rho$, $S$, $q_x$, $\lambda$/$R$, $t^*$) y el protocolo
estadístico son los mismos, de modo que las dos secciones se leen en paralelo. Lo que cambia entre
ellas es sólo el invariante de la familia: $\phi$ fija allá, $\mu_s$ igualado aquí. Conviene que el
texto lo diga en una frase al abrir la sección, porque de ahí sale todo lo demás.

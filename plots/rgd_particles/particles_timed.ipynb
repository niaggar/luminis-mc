## Funciones de análisis

Pipeline por réplica: se leen los dos detectores, se **unen** (`load_stitched`),
se calcula el enhancement $E = I_{\rm coh}/I_{\rm incoh}$, y se lleva la cola a 1
restando el fondo (`tail_baseline`). Sobre eso se calculan las métricas
$\eta(0)$ (`peak_height`) y el ancho FWHM en $q$ (`fwhm_q`). Las estadísticas
sobre réplicas (media y desviación estándar) las hacen `profile_stats` y
`metric_stats`.

```python
def lstar_teoria(sweep, key):
    """l* teorico guardado en los parametros de la simulacion."""
    p = sweep[key].params_flat
    return p.get("transport_mean_free_path", p.get("lstar_sim", 0.0))


def load_stitched(lstar, th1, g1, th2, g2, key, channel=""):
    """Une los perfiles de los dos detectores (fino + grueso) en un solo eje.

    Chequea el solape entre ambos: si difieren mas de 5% avisa para revisar
    la normalizacion de angulo solido. Devuelve (theta, q, g) con
    q = k * l* * theta."""
    theta_1 = th1[-1]            # fin de la ventana fina
    lo = 0.9 * theta_1           # inicio real de det_2 (solape)

    ov = (th2 >= lo) & (th2 <= theta_1)
    if ov.any():
        g1_on2 = np.interp(th2[ov], th1, g1)
        resid = np.abs(g2[ov] - g1_on2)
        if np.max(resid) > 0.05:
            print(f"[stitch WARN] {key} ch={channel}: max solape "
                  f"{np.max(resid):.3f} -- revisar normalizacion angulo solido")

    keep2 = th2 > theta_1
    theta = np.concatenate([th1, th2[keep2]])
    g = np.concatenate([g1, g2[keep2]])
    q = K_MEDIUM * lstar * theta
    return theta, q, g


def tail_baseline(q, y, q_min=25.0, n_min=5, frac=0.85):
    """Fondo en la cola, inmune a ventanas vacias.
    Usa q>q_min si hay puntos suficientes; si no, cae al tramo alto
    del q realmente disponible para ese radio."""
    q = np.asarray(q); y = np.asarray(y)
    m = q > q_min
    if np.count_nonzero(m) >= n_min:
        return np.median(y[m])
    q_cut = frac * np.nanmax(q)          # ultimo ~15 % del q disponible
    m = q >= q_cut
    if np.count_nonzero(m) < n_min:
        idx = np.argsort(q)[-n_min:]     # los n_min puntos de mayor q
        return np.median(y[idx])
    return np.median(y[m])


def profile_one(sweep, key, phi_index, channel, time_index=0, basis=linear):
    """Perfil E(q) de UNA replica, con la baseline llevada a 1.

    Ojo: en base circular el reductor es el promedio azimutal, asi que
    phi_index no tiene efecto en ese caso."""
    reducer = phi_cut(phi_index) if basis == linear else azimuthal_average
    sw = sweep[key]
    p1 = cbs_profiles(sw.processed_cbs("farfield_cbs_1"), basis=basis,
                      time_index=time_index, reduce=reducer)
    p2 = cbs_profiles(sw.processed_cbs("farfield_cbs_2"), basis=basis,
                      time_index=time_index, reduce=reducer)
    E1 = (p1.coherent[channel] + EPS) / (p1.incoherent[channel] + EPS)
    E2 = (p2.coherent[channel] + EPS) / (p2.incoherent[channel] + EPS)
    theta, q, E = load_stitched(lstar_teoria(sweep, key),
                                p1.theta, E1, p2.theta, E2, key, channel)
    return theta, q, E - tail_baseline(q, E) + 1.0


def profile_stats(sweep, grouped_rad, phi_index, channel, time_index=0, basis=linear):
    """Media y desviacion estandar (ddof=1) del perfil sobre las replicas."""
    curves = []
    for key in grouped_rad.replicas_keys:
        theta, q, E = profile_one(sweep, key, phi_index, channel,
                                  time_index=time_index, basis=basis)
        curves.append(E)
    C = np.vstack(curves)
    return theta, q, C.mean(axis=0), C.std(axis=0, ddof=1)


def peak_height(q, E, n=3):
    """E(0) estimado como mediana de los primeros n bins (robusto al ruido del apice)."""
    return np.median(E[:n])


def fwhm_q(q, E, n=3):
    """Ancho total a mitad de altura en unidades de q, por interpolacion lineal."""
    E0 = peak_height(q, E, n)
    half = 1.0 + 0.5 * (E0 - 1.0)
    below = np.where(E < half)[0]
    if len(below) == 0:
        return np.nan
    i = below[0]                                  # primer bin bajo el semi-maximo
    q_half = q[i-1] + (half - E[i-1]) * (q[i] - q[i-1]) / (E[i] - E[i-1])
    return 2.0 * q_half                            # cono simetrico en theta


def metric_stats(sweep, grouped_rad, phi_index, channel, fn, basis=linear):
    """(media, std) de una metrica fn(q, E) evaluada replica a replica."""
    vals = []
    for key in grouped_rad.replicas_keys:
        _, q, E = profile_one(sweep, key, phi_index, channel, basis=basis)
        vals.append(fn(q, E))
    return np.mean(vals), np.std(vals, ddof=1)


def mirror(q, *ys):
    """Espejo -q..+q para presentacion estilo Iwai (valido: I(th,phi)=I(th,phi+pi))."""
    return (np.r_[-q[::-1], q],) + tuple(np.r_[y[::-1], y] for y in ys)


def rgd_number(rad_um):
    """2x|m-1| con x = k_medio * R. RGD estricto si << 1."""
    x = K_MEDIUM * rad_um
    return 2.0 * x * abs(M_REL - 1.0)
```

## Funciones de graficado

Solo hay dos tipos de figura de perfiles y cada uno se usa para ambas
polarizaciones, así que se definen una vez:

- `plot_perfiles_pareja` — dos paneles (un canal cada uno), todas las tallas superpuestas, $\phi = 0$.
- `plot_rejilla_phi` — rejilla 2×3 (un panel por talla) comparando $\phi = 0°, 45°, 90°$.

```python
def banda(ax, q, m, s, color, **kw):
    """Curva media con banda +-1 sigma, espejada a -q..+q."""
    qs, ms, ss = mirror(q, m, s)
    ax.plot(qs, ms, color=color, **kw)
    ax.fill_between(qs, ms - ss, ms + ss, color=color, alpha=0.20, lw=0)


def lineas_referencia(ax):
    ax.axhline(1.0, ls="--", c="gray", lw=0.7)
    ax.axhline(2.0, ls="--", c="gray", lw=0.7)


def leyenda_inferior(fig, ax, ncol=6):
    fig.legend(*ax.get_legend_handles_labels(),
               ncol=ncol, loc="outside lower center",
               frameon=False, columnspacing=1.2, handlelength=1.3)


def plot_perfiles_pareja(sweep, basis, paneles, fname):
    """Dos paneles lado a lado, uno por canal, phi = 0 (X-scan).

    paneles: [(canal, titulo), (canal, titulo)]"""
    apply(width_frac=1.0)
    fig, axs = plt.subplots(1, 2, figsize=(TEXTWIDTH_IN, 0.6*TEXTWIDTH_IN), sharey=True)

    for c, g_ in zip(COL, grouped_data):
        for ax, (ch, _) in zip(axs, paneles):
            _, q, m, s = profile_stats(sweep, g_, 0, ch, basis=basis)
            banda(ax, q, m, s, color=c, label=g_.name)

    for ax, (_, titulo) in zip(axs, paneles):
        ax.set_xlim(-10, 10)
        ax.set_ylim(0.95, 2.05)
        ax.set_xlabel(r"$q$")
        ax.set_title(titulo, loc="left")
        ax.grid(False)
        lineas_referencia(ax)

    axs[0].set_ylabel(r"$\eta$")
    leyenda_inferior(fig, axs[0])
    fig.savefig(FIGDIR / fname)
    plt.show()


PHI_STYLE = [(0, COL[0], "-",  r"$\phi=0^\circ$"),
             (1, COL[1], "--", r"$\phi=45^\circ$"),
             (2, COL[2], "-.", r"$\phi=90^\circ$")]

def plot_rejilla_phi(sweep, basis, channel, etiqueta, fname):
    """Rejilla 2x3 (un panel por radio) con los cortes phi = 0, 45, 90."""
    apply(width_frac=1.0)
    fig, axes = plt.subplots(2, 3, figsize=(TEXTWIDTH_IN, 0.62*TEXTWIDTH_IN),
                             sharex=True, sharey=True)

    for ax, g_ in zip(axes.flat, grouped_data):
        for pi, c, ls, lb in PHI_STYLE:
            _, q, m, s = profile_stats(sweep, g_, pi, channel, basis=basis)
            banda(ax, q, m, s, color=c, ls=ls, label=lb)

        ax.set_title(g_.name, loc="left")
        ax.set_xlim(-8, 8)
        ax.set_ylim(0.95, 2.05)
        ax.grid(False)
        lineas_referencia(ax)

    for ax in axes[1]:    ax.set_xlabel(r"$q$")
    for ax in axes[:, 0]: ax.set_ylabel(r"$\eta$")

    axes[0, 0].text(0.04, 0.90, etiqueta, transform=axes[0, 0].transAxes,
                    ha="left", va="top")
    leyenda_inferior(fig, axes[0, 0])
    fig.savefig(FIGDIR / fname)
    plt.show()
```

# Ajuste del cono difusivo

## 1. Modelo del cono difusivo y ajuste por replica

```python
def cone_shape(x, b=B):
    """Cono difusivo normalizado (Akkermans et al. 1986). C(0) = 1."""
    x = np.maximum(np.asarray(x, float), 1e-9)
    return (1.0/(1.0 + 2.0*b)) / (1.0 + x)**2 * (1.0 + (1.0 - np.exp(-2.0*b*x)) / x)


def fit_cone(q, E, q_max=6.0):
    """Ajusta E(q) = 1 + A * C(r*q).

    q  = k * l*_teo * theta   (reducido con el l* teorico)
    r  = l*_med / l*_teo      (parametro libre: el reescalado del ancho)
    A  = amplitud del pico    (A = eta(0) - 1)

    Devuelve (r, A). No hay ninguna constante calibrada aqui: r es la
    razon cruda entre el ancho medido y el ancho difusivo esperado.
    """
    q = np.asarray(q, float); E = np.asarray(E, float)
    m = (q < q_max) & np.isfinite(E)
    popt, _ = curve_fit(lambda qq, r, A: 1.0 + A * cone_shape(r * qq),
                        q[m], E[m], p0=[1.0, 1.0],
                        bounds=([0.02, 0.0], [50.0, 1.2]), maxfev=20000)
    return popt[0], popt[1]
```

## 2. Inversion: tabla unica de observables

Por cada radio y cada replica se ajusta el cono. De ahi salen los dos
observables independientes:

- **transporte** -> `lstar_med` (del **ancho** del cono)
- **dispersion simple** -> `rho = 2 - eta(0)` (de la **altura** del cono)

```python
def invert(sweep, basis=linear, channel="co", phi=0, q_max=6.0):
    """Ajusta el cono replica a replica -> l*_med, rho y sus sigmas."""
    rows = []
    for g_ in grouped_data:
        lteo = lstar_teoria(sweep, g_.replicas_keys[0])
        rr, AA, EE = [], [], []
        for key in g_.replicas_keys:
            _, q, E = profile_one(sweep, key, phi, channel, basis=basis)
            try:
                r, A = fit_cone(q, E, q_max=q_max)
            except Exception as err:
                print(f"[fit FAIL] {key}: {err}")
                continue
            rr.append(r); AA.append(A); EE.append(peak_height(q, E))
        rr, AA, EE = np.asarray(rr), np.asarray(AA), np.asarray(EE)
        rows.append({
            "R [nm]":    g_.radius * 1e3,
            "lstar_teo": lteo,
            "lstar_raw": rr.mean() * lteo,          # sin calibrar
            "s_raw":     rr.std(ddof=1) * lteo,
            "ratio":     rr.mean(),                 # l*_raw / l*_teo
            "s_ratio":   rr.std(ddof=1),
            "eta0":      EE.mean(),
            "rho":       2.0 - EE.mean(),           # peso de single scattering
            "s_rho":     EE.std(ddof=1),
            "A":         AA.mean(),
            "2x|m-1|":   rgd_number(g_.radius),
            "w/l*":      LASER_RADIUS_UM / lteo,
        })
    return pd.DataFrame(rows)


# --- caso principal: polarizacion lineal, canal co ---
df = invert(sweep_data_lineal, basis=linear, channel="co", phi=0)
df.round(4)
```

## 3. Constante de calibracion

`ratio = l*_med / l*_teo` sale **plana** en el rango donde la lineshape es
universal: eso valida el escalado `theta_FWHM ~ 1/l*`. Lo unico que falla es
el **prefactor**, porque el cono real no es el difusivo puro (eta(0) < 2:
domina la dispersion de orden bajo). Se calibra **una sola constante global**
`beta` sobre el rango valido, no una particula de referencia.

```python
def calibrar(d):
    """Calibra beta sobre el rango RGD estricto (2x|m-1| < 1) y agrega las
    columnas lstar_med, s_lstar y err_% al DataFrame. Devuelve (beta, s_beta)."""
    cal = d["2x|m-1|"] < 1.0
    beta   = d.loc[cal, "ratio"].mean()
    s_beta = d.loc[cal, "ratio"].std(ddof=1)
    d["lstar_med"] = d["lstar_raw"] / beta
    d["s_lstar"]   = d["s_raw"]     / beta
    d["err_%"]     = 100.0 * (d["lstar_med"] / d["lstar_teo"] - 1.0)
    return beta, s_beta


beta, s_beta = calibrar(df)
CAL = df["2x|m-1|"] < 1.0

print(f"radios de calibracion : {df.loc[CAL, 'R [nm]'].tolist()} nm")
print(f"beta = {beta:.4f} +/- {s_beta:.4f}   (dispersion {100*s_beta/beta:.1f} %)")
print(f"  -> el cono medido es {1/beta:.2f}x mas ancho que el difusivo (b = {B})")

df[["R [nm]", "lstar_teo", "lstar_med", "s_lstar", "err_%",
    "eta0", "rho", "2x|m-1|"]].round(3)
```

## 5. FIG B -- Plano de observables independientes

Ambos ejes son **medidos**. El radio no es un observable directo del CBS:
corre a lo largo de la curva. Para recuperar R hay que **combinar** los dos
observables (o conocer la densidad de forma independiente).

## Ley temporal del estrechamiento

El cono resuelto en tiempo se estrecha como $\theta_{\rm FWHM} \propto 1/\sqrt{t}$
(Vreeker *et al.*, Phys. Lett. A **132**, 51, 1988; Tourin *et al.*, PRL **79**,
3637, 1997 — pendiente medida $0.49 \approx 1/2$). El origen es difusivo: a
tiempo $t$ solo contribuyen caminos de longitud $ct$, cuyos extremos estan
separados $\sim\sqrt{D t}$, y el exceso coherente es una gaussiana
$\propto \exp(-D\,t\,k^2\theta^2)$ con $D = c\,\ell^*/3$.

En las unidades reducidas del notebook ($q = k\ell^*\theta$, $s = t/\tau^*$ con
$\tau^* = \ell^*/c$) la ley queda **universal**, independiente del radio:

$$\Delta q_{\rm FWHM}^2(s) \;=\; \frac{12\ln 2}{s} \;\approx\; \frac{8.32}{s}.$$

Se verifica con tres chequeos: **(a)** pendiente $-1/2$ de
$\Delta q_{\rm FWHM}(s)$ en log-log, **(b)** plateau de la curva compensada
$s\,\Delta q^2$ en $12\ln 2$ (su desviacion mide $D_{\rm med}/D_{\rm teo}$),
**(c)** $\eta(0,s) \to 2$ al gatear los ordenes bajos (el deficit $\rho$ vive
en los primeros bines: single/low-order scattering).

```python
# FWHM y altura del pico por bin temporal, replica a replica.
# Bines log-espaciados (la ley es una potencia); se descartan por bin las
# replicas donde el pico ya no cruza el semi-maximo (fwhm_q -> nan).
time_bins = np.unique(np.round(np.geomspace(1, 90, 25)).astype(int))
s_vals    = t_of_bin(time_bins)

temporal = {}    # nombre del radio -> {"s", "dq": (m, s), "e0": (m, s)}
for g_ in grouped_use:
    dq, e0 = [], []
    for key in g_.replicas_keys:
        dq_rep, e0_rep = [], []
        for it in time_bins:
            _, q, E = profile_one(sweep_data_circular, key, 0, "cross",
                                  time_index=int(it), basis=circular)
            dq_rep.append(fwhm_q(q, E))
            e0_rep.append(peak_height(q, E))
        dq.append(dq_rep); e0.append(e0_rep)
    dq, e0 = np.array(dq), np.array(e0)
    temporal[g_.name] = {
        "s":  s_vals,
        "dq": (np.nanmean(dq, axis=0), np.nanstd(dq, axis=0, ddof=1)),
        "e0": (e0.mean(axis=0),        e0.std(axis=0, ddof=1)),
    }
```

```python
apply(width_frac=1.0)
fig, (a1, a2, a3) = plt.subplots(1, 3, figsize=(TEXTWIDTH_IN, 0.4*TEXTWIDTH_IN))

C_TEO  = 12.0 * np.log(2.0)          # s * dq^2 difusivo (D = c l*/3)
s_line = np.geomspace(s_vals[0], s_vals[-1], 100)

for c, g_ in zip(COL, grouped_use):
    d = temporal[g_.name]
    s, (dq_m, dq_s), (e0_m, e0_s) = d["s"], d["dq"], d["e0"]

    a1.errorbar(s, dq_m, dq_s, fmt="o", ms=3, capsize=2, ls="none",
                color=c, label=g_.name)
    a2.errorbar(s, s * dq_m**2, 2.0 * s * dq_m * dq_s, fmt="o", ms=3,
                capsize=2, ls="none", color=c)
    a3.errorbar(s, e0_m, e0_s, fmt="o-", ms=3, lw=0.8, capsize=2, color=c)

# (a) estrechamiento log-log con la ley universal
a1.plot(s_line, np.sqrt(C_TEO / s_line), "--", c="gray", lw=0.8, zorder=0)
a1.set_xscale("log"); a1.set_yscale("log")
a1.set_xlabel(r"$s = t/\tau^*$")
a1.set_ylabel(r"$\Delta q_{\mathrm{FWHM}}$")
a1.set_title("(a)", loc="left")
a1.annotate(r"slope $=-1/2$", (0.05, 0.08), xycoords="axes fraction",
            fontsize=8, color="gray")

# (b) curva compensada: plateau = regimen difusivo
a2.axhline(C_TEO, ls="--", c="gray", lw=0.8, zorder=0)
a2.set_xscale("log")
a2.set_xlabel(r"$s = t/\tau^*$")
a2.set_ylabel(r"$s\,\Delta q_{\mathrm{FWHM}}^2$")
a2.set_title("(b)", loc="left")
a2.annotate(r"$12\ln 2$", (0.05, 0.90), xycoords="axes fraction",
            fontsize=8, color="gray")

# (c) recuperacion del enhancement al gatear ordenes bajos
a3.axhline(2.0, ls="--", c="gray", lw=0.7)
a3.axhline(1.0, ls=":",  c="gray", lw=0.7)
a3.set_xscale("log")
a3.set_xlabel(r"$s = t/\tau^*$")
a3.set_ylabel(r"$\eta(0)$")
a3.set_ylim(0.95, 2.15)
a3.set_title("(c)", loc="left")

for ax in (a1, a2, a3):
    ax.grid(False)

fig.legend(*a1.get_legend_handles_labels(),
           ncol=len(grouped_use), loc="outside lower center",
           frameon=False, columnspacing=1.6, handlelength=1.6)
fig.savefig(FIGDIR / "cbs_temporal_narrowing.pdf")
plt.show()
```

### Pendiente y coeficiente de difusion

Ajuste log-log en la ventana difusiva `S_FIT` (se excluyen los primeros bines,
donde el transporte aun no es difusivo y domina la dispersion de orden bajo, y
los ultimos, dominados por ruido de conteo). De la curva compensada sale el
coeficiente de difusion relativo sin ningun parametro libre:

$$\frac{D_{\rm med}}{D_{\rm teo}} = \frac{12\ln 2}{\langle s\,\Delta q^2\rangle},
\qquad D_{\rm teo} = \tfrac{1}{3}\,c\,\ell^*.$$

Esto separa $D$ de $\ell^*$ (el estacionario solo da el producto via el ancho
del cono integrado), que es justamente la informacion extra del eje temporal.

```python
S_FIT = (3.0, 30.0)     # ventana difusiva en s = t/tau*

rows = []
for g_ in grouped_use:
    d = temporal[g_.name]
    s, dq_m = d["s"], d["dq"][0]
    m = (s >= S_FIT[0]) & (s <= S_FIT[1]) & np.isfinite(dq_m)

    p, _  = np.polyfit(np.log(s[m]), np.log(dq_m[m]), 1)   # esperado: -1/2
    C_med = np.median(s[m] * dq_m[m]**2)                   # esperado: 12 ln2

    rows.append({
        "R [nm]":     g_.radius * 1e3,
        "pendiente":  p,
        "s*dq^2":     C_med,
        "D_med/D_teo": C_TEO / C_med,
        "n_bins":     int(np.count_nonzero(m)),
    })

df_t = pd.DataFrame(rows)
df_t.to_csv(FIGDIR / "cbs_temporal_law.csv", index=False)
print(f"esperado: pendiente = -0.5,  s*dq^2 = 12 ln2 = {C_TEO:.3f}")
df_t.round(3)
```
